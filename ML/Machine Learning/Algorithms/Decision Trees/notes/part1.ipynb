Decision Trees - Part 1: Introduction and Intuition
Decision Trees

A Decision Tree is a supervised machine learning algorithm used for both:

Classification
Regression

It predicts the output by learning a series of decision rules from the training data. These rules are organized in a tree-like structure, where each internal node asks a question about a feature, each branch represents the answer to that question, and each leaf node contains the final prediction.

Instead of learning a mathematical equation like Linear Regression or Logistic Regression, a Decision Tree learns a sequence of if-else conditions.

Real-Life Intuition

Every day, humans make decisions using a series of questions.

Suppose you are deciding whether to play cricket.

Is it raining?

        Yes
         |
      Stay Home

        No
         |
Is it weekend?

   Yes        No
    |          |
 Play      Go to Work

Notice that each question reduces uncertainty until a final decision is reached.

A Decision Tree works in exactly the same way.

Instead of asking random questions, it automatically learns the best questions from the training data.

Why Are Decision Trees Powerful?

Many machine learning algorithms try to fit equations.

For example:

Linear Regression

y = mx + c

Logistic Regression

P(y)=1 / (1+e^-z)

SVM

Find the best separating hyperplane.

Decision Trees are different.

They do not learn equations.

Instead, they learn rules like:

IF Age < 30
AND Salary > 50K
THEN Buy = Yes

These rules are easy for humans to understand.

This is why Decision Trees are considered one of the most interpretable machine learning algorithms.

Types of Decision Trees

Decision Trees can solve two types of machine learning problems.

1. Classification Tree

Used when the target variable is categorical.

Example:

Spam

Not Spam

or

Pass

Fail

or

Yes

No

Output is a class label.

2. Regression Tree

Used when the target variable is continuous.

Example:

Predict

House Price = ₹52,30,000

or

Temperature = 32.7°C

Instead of predicting a class, it predicts a numerical value.

Why Is It Called a "Tree"?

The structure resembles an upside-down tree.

               Root

              Age?

          /          \

      Student?     Salary?

      /      \      /     \

    Buy   Don't   Buy   Don't
Top node → Root
Middle nodes → Decision Nodes
Bottom nodes → Leaf Nodes

Information flows from the root to a leaf, just like following branches of a tree.

Characteristics of Decision Trees

A Decision Tree:

Is a supervised learning algorithm.
Can perform both classification and regression.
Learns decision rules from data.
Does not assume any relationship between features and target (non-parametric).
Handles both numerical and categorical features (implementation support varies).
Is easy to visualize and interpret.
Can capture non-linear relationships.
Can model interactions between features automatically.
Advantages
Easy to understand and explain.
No need for feature scaling or normalization.
Works with both numerical and categorical data.
Captures non-linear patterns.
Can handle multi-class classification naturally.
Performs feature selection implicitly by choosing informative splits.
Requires little data preprocessing.
Disadvantages
Prone to overfitting if grown too deep.
Small changes in data can produce a very different tree (high variance).
Can be biased toward features with many unique values if not carefully controlled.
Usually less accurate than ensemble methods like Random Forests or Gradient Boosting.
Greedy split decisions may not lead to the globally optimal tree.
How Does a Decision Tree Learn?

At a high level, the algorithm repeatedly follows this process:

Start with all training data
        │
        ▼
Choose the best feature and threshold to split the data
        │
        ▼
Divide the data into smaller groups
        │
        ▼
Repeat the process for each new group
        │
        ▼
Stop when a stopping criterion is met
        │
        ▼
Assign a prediction to each leaf node

The key question is:

How does the algorithm decide what the "best" split is?

This is the core of Decision Trees, and it leads us to the next topic: Purity, Impurity, Entropy, Gini Impurity, and Information Gain.

Key Takeaways
A Decision Tree is a supervised learning algorithm for both classification and regression.
It predicts by asking a sequence of learned if-else questions.
The algorithm builds the tree by repeatedly splitting the data into more homogeneous groups.
Classification Trees predict categories, while Regression Trees predict continuous values.
Decision Trees are highly interpretable but can easily overfit without proper controls.
The effectiveness of a tree depends on selecting the best split at each node, which requires understanding purity and impurity.
