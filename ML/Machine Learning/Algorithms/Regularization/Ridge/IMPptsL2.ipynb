# 3. Bias-Variance Tradeoff


This states that the bias and the variance of the dataset should be well maintained such that the loss function is minimized for a given value of alpha/lambda

Finding the "Sweet Spot"Total model error is essentially the sum of \(\text{Bias}^2 + \text{Variance} + \text{Irreducible Error}\). Because lowering bias usually increases variance (and vice versa), calculating these values allows you to track your model along the Bias-Variance Tradeoff curve to find the optimal point of lowest overall error.
