# 1. Introduction

Support Vector Machine (SVM) is one of the most influential supervised machine learning algorithms ever developed.

It is primarily used for:

- Binary Classification
- Multi-class Classification
- Regression (Support Vector Regression)
- Outlier Detection

Unlike many algorithms that simply try to classify data correctly, SVM has a different philosophy:

> Instead of finding **any** decision boundary, SVM searches for the **best possible decision boundary**.

This single idea is what makes SVM unique.

---

# 2. Why Was SVM Invented?

Imagine you have two classes:

```
○ = Apples

● = Oranges
```

```
○ ○ ○ ○

        ● ● ● ●
```

Many algorithms can separate these two classes.

For example:

```
○ ○ ○ │ ● ● ●
○ ○ ○ │ ● ● ●
○ ○ ○ │ ● ● ●
```

or

```
○ ○ │ ○ ● ●
○ ○ │ ● ● ●
○ ○ │ ● ● ●
```

or

```
○ │ ○ ○ ● ●
○ │ ○ ● ● ●
○ │ ○ ● ● ●
```

All of these classify the training data correctly.

Which one is the best?

Traditional classifiers don't explicitly answer this question.

SVM does.

---

# 3. The Central Idea

SVM asks:

> Which decision boundary leaves the largest possible gap between the two classes?

That gap is called the **Margin**.

Instead of simply separating data,

SVM maximizes the margin.

This leads to better generalization.

---

# 4. Intuition

Imagine you are building a road between two villages.

```
Village A

○ ○ ○ ○


Village B

● ● ● ●
```

Would you build the road

close to Village A?

or

close to Village B?

Probably neither.

You would place it in the middle.

Why?

Because it maximizes safety.

SVM thinks exactly the same way.

The safest decision boundary is usually the one furthest away from both classes.

---

# 5. What is a Hyperplane?

A hyperplane is simply the surface that separates two classes.

In different dimensions:

## One Dimension

A hyperplane is a point.

```
------|------
```

---

## Two Dimensions

A hyperplane is a line.

```
○ ○ ○

-----------

● ● ●
```

---

## Three Dimensions

A hyperplane becomes a plane.

```
Plane separating objects
```

---

## Higher Dimensions

A hyperplane becomes an (n−1)-dimensional surface.

Humans cannot visualize this,

but mathematically it behaves exactly the same.

---

# 6. Hyperplane Equation

For two features,

the decision boundary is

\[
w_1x_1+w_2x_2+b=0
\]

where

- x = feature vector
- w = weights
- b = bias

You may recognize this equation.

It is almost identical to Logistic Regression.

So what makes SVM different?

The objective function.

Not the decision boundary equation.

---

# 7. Decision Boundary

The hyperplane divides the feature space into two regions.

```
Ham

○ ○ ○

-------------

● ● ●

Spam
```

Every new sample is classified according to which side of the hyperplane it falls.

# 10. Margin

The margin is the distance between

the decision boundary

and

the nearest data points.

```
○ ○ ○

----------------

● ● ●
```

Imagine drawing two parallel lines touching the closest points from each class.

The distance between them

is called

the Margin.

---

# 11. Maximum Margin Principle

This is the heart of SVM.

Among all separating hyperplanes,

choose the one with

the largest margin.

Why?

Because larger margins usually produce better generalization.

---

# 12. Why Larger Margin Helps

Suppose your data contains measurement noise.

Training data

```
○ ○ ○

-----------

● ● ●
```

New data

```
○ ○ ○

   ○

-----------

● ● ●
```

If the boundary is too close,

the new sample may be misclassified.

A larger margin provides a safety buffer.

This is one reason SVM performs well on unseen data.

---

# 13. Support Vectors

Surprisingly,

most training points are irrelevant.

Only the closest samples matter.

Example

```
○ ○ ○ ○

   ○

------------

   ●

● ● ● ●
```

The highlighted samples are

Support Vectors.

They define the margin.

Move any other point,

the decision boundary barely changes.

Move a support vector,

the boundary changes.

Hence the name

Support Vector Machine.

Everything depends on the support vectors.

The remaining samples mostly have no influence.

# 15. Hard Margin SVM

Suppose every point is perfectly separable.

```
○ ○ ○

--------------

● ● ●
```

Then

SVM insists

every point must be correctly classified.

No mistakes allowed.

This is called

Hard Margin SVM.

---

Mathematically

Every sample must satisfy

\[
y_i(w^Tx_i+b)\ge1
\]

We'll derive this equation in Part 2.

One outlier makes perfect separation impossible.

# 18. Key Takeaways

✔ SVM searches for the best decision boundary.

✔ "Best" means maximum margin.

✔ Only support vectors influence the model.

✔ Hyperplane separates the feature space.

✔ Hard Margin requires perfectly separable data.

✔ Larger margins generally improve generalization.

✔ The optimization objective distinguishes SVM from Logistic Regression.

# Interview Questions

### Q1. Why doesn't SVM choose any separating hyperplane?

Because infinitely many hyperplanes can separate the data. SVM selects the one with the maximum margin, which generally provides better generalization.

---

### Q2. What is a support vector?

The training samples closest to the decision boundary. They alone determine the position of the optimal hyperplane.

---

### Q3. Why are they called support vectors?

Because they "support" or define the optimal separating hyperplane. Removing a support vector changes the boundary much more than removing other points.

---

### Q4. What is a margin?

The distance between the decision boundary and the nearest training samples from each class.

---

### Q5. Why does maximizing the margin improve performance?

A larger margin makes the classifier less sensitive to small variations and noise in unseen data, reducing the risk of overfitting.

# Common Misconceptions

❌ SVM uses all training samples equally.

✔ Only the support vectors have a direct influence on the final decision boundary.

---

❌ SVM always uses kernels.

✔ Linear SVM uses no kernel trick.

---

❌ Hyperplane means a plane.

✔ In machine learning, a hyperplane is the general decision surface in any number of dimensions.

# Cheat Sheet

• Goal → Find the maximum-margin hyperplane.

• Decision Boundary → Hyperplane.

• Closest Points → Support Vectors.

• Margin → Distance from hyperplane to support vectors.

• Hard Margin → No classification errors allowed.

• Soft Margin → Allows some classification errors (covered in Part 3).
