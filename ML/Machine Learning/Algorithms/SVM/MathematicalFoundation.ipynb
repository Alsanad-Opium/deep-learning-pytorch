# 2. Feature Space

Suppose every SMS message is represented using two features.

Example

| Message | x₁ | x₂ |
|---------|---:|---:|
| A | 2 | 4 |
| B | 5 | 6 |
| C | 3 | 7 |

Each sample becomes a point.

```

x₂

^

|

7 • C

|

6 • B

|

4 • A

+---------------------->

x₁

```

The collection of all possible feature values is called the **Feature Space**.

Every machine learning classifier works by dividing this space.

# 3. Vectors

A point

```

(3,5)

```

can also be written as

\[
\mathbf{x}
=
\begin{bmatrix}
3\\
5
\end{bmatrix}
\]

This is called a **Feature Vector**.

Generally

\[
x=
[x_1,x_2,\ldots,x_n]
\]

# 4. Weight Vector

Suppose

\[
w=
\begin{bmatrix}
2\\
1
\end{bmatrix}
\]

The weight vector tells us

- which direction the decision boundary faces
- how important each feature is

Unlike Linear Regression,

the weights here define the orientation of the separating hyperplane.

# 5. Dot Product

This is probably the most important mathematical operation in SVM.

For two vectors

\[
a=
[a_1,a_2]
\]

and

\[
b=
[b_1,b_2]
\]

their dot product is

\[
a\cdot b
=
a_1b_1+a_2b_2
\]

Example

\[
(2,3)\cdot(4,5)
=
2\times4+3\times5
=
23
\]

---

# 6. Geometric Meaning of Dot Product

A much deeper interpretation is

\[
a\cdot b
=
||a||
||
b||
\cos\theta
\]

where

- θ = angle between vectors

If

θ=0°

Dot Product is maximum.

If

θ=90°

Dot Product becomes zero.

If

θ=180°

Dot Product becomes negative.

---

Why is this useful?

Because the dot product tells us

**how aligned two vectors are.**

This becomes crucial for SVM.

# 7. Hyperplane Equation

The hyperplane is defined by

\[
w^Tx+b=0
\]

Let's understand every term.

---

## x

Feature vector

Example

\[
x=
\begin{bmatrix}
3\\
2
\end{bmatrix}
\]

---

## w

Weight vector

Example

\[
w=
\begin{bmatrix}
4\\
1
\end{bmatrix}
\]

---

## b

Bias

Moves the hyperplane left or right.

---

The decision function becomes

\[
f(x)
=
w^Tx+b
\]

---

# 8. What Does the Equation Mean?

Suppose

\[
2x_1+x_2-6=0
\]

This is simply a straight line.

Everything satisfying

\[
2x_1+x_2-6=0
\]

lies exactly on the decision boundary.

---

Points giving

Positive values

lie on one side.

Negative values

lie on the other.

Example

Suppose

\[
2(4)+2-6=4
\]

Positive

↓

Class 1

Now

\[
2(1)+1-6=-3
\]

Negative

↓

Class 2

---

# 9. Decision Function

The classifier predicts using

\[
f(x)=w^Tx+b
\]

Prediction becomes

\[
\hat y
=
sign(f(x))
\]

Meaning

if

\[
f(x)>0
\]

Class +1

if

\[
f(x)<0
\]

Class -1

Notice

SVM does **not**

predict probabilities.

It predicts which side of the hyperplane the sample lies on.

---

# 10. Signed Distance

One of the most beautiful ideas in SVM.

The distance of a point from the hyperplane is

\[
\frac{w^Tx+b}{||w||}
\]

Notice

We divide by

\[
||w||
\]

because larger weight vectors would otherwise artificially inflate the score.

---

# 11. Vector Norm

The norm is simply

vector length.

For

\[
w=(3,4)
\]

Length

\[
||w||
=
\sqrt{3^2+4^2}
=
5
\]

---

# 12. Why Normalize?

Imagine

Hyperplane A

Weight

\[
w=(2,1)
\]

Hyperplane B

Weight

\[
w=(20,10)
\]

Both represent exactly the same line.

Without normalization

the second appears to have larger distances.

That's incorrect.

Hence

divide by

\[
||w||
\]

---

# 13. Functional Margin

For a training sample

\[
(x_i,y_i)
\]

the functional margin is

\[
y_i(w^Tx_i+b)
\]

Interpretation

Large positive

↓

Correctly classified

Near zero

↓

Very close to boundary

Negative

↓

Wrong side

---

Example

Suppose

True class

\[
y=1
\]

Decision value

\[
w^Tx+b=5
\]

Functional margin

\[
1\times5=5
\]

Excellent.

Now

Suppose

\[
y=-1
\]

Decision value

\[
-4
\]

Margin

\[
(-1)(-4)=4
\]

Also correct.

---

# 14. Geometric Margin

Functional margin depends on the scale of

\[
w
\]

So we normalize.

The geometric margin becomes

\[
\gamma
=
\frac{y_i(w^Tx_i+b)}
{||w||}
\]

This is the true perpendicular distance from the hyperplane.

---

# 15. The SVM Goal

Among all hyperplanes

find one with

largest geometric margin.

This gives

Maximum Margin Classifier.

---

# 16. Why Does Margin Equal 2/||w||?

Support vectors satisfy

\[
w^Tx+b=1
\]

and

\[
w^Tx+b=-1
\]

These are two parallel planes.

Distance between them

is

\[
\frac{2}{||w||}
\]

This is the margin.

Notice something amazing.

Margin

is inversely proportional to

\[
||w||
\]

Large weights

↓

Small margin

Small weights

↓

Large margin

---

# 17. Therefore

Maximizing

\[
\frac{2}{||w||}
\]

is equivalent to

minimizing

\[
||w||
\]

For easier optimization,

we minimize

\[
\frac12||w||^2
\]

instead.

Why square it?

- Easier derivatives
- Convex objective
- Same minimum

---

# 18. Optimization Problem

The complete hard-margin optimization problem becomes

Minimize

\[
\frac12||w||^2
\]

Subject to

\[
y_i(w^Tx_i+b)\ge1
\]

This is the heart of SVM.

Every SVM solver is ultimately solving this optimization problem.

---

# 19. Why the Constraint ≥1?

Suppose

Positive class

\[
w^Tx+b=5
\]

Correct.

Now

Suppose

\[
w^Tx+b=0.0001
\]

Still correct,

but extremely close to the boundary.

SVM wants points to stay

outside the margin.

Hence

\[
y_i(w^Tx_i+b)\ge1
\]

instead of merely

>0

---

# 20. Support Vectors Revisited

Support vectors satisfy

\[
y_i(w^Tx_i+b)=1
\]

Exactly.

All other samples satisfy

\[
>1
\]

Therefore

only support vectors touch the margin.

These are the only samples that influence the optimal solution.

# Interview Questions

### Q1. Why do we minimize ||w||² instead of maximizing the margin directly?

Because the margin is inversely proportional to \(||w||\). Minimizing \(||w||^2\) is mathematically equivalent to maximizing the margin and results in a smooth, convex optimization problem.

---

### Q2. Why do we square the norm?

Squaring removes the square root, simplifies differentiation, and preserves the same optimum because the square function is monotonically increasing for non-negative values.

---

### Q3. Why is the decision function not a probability?

Unlike Logistic Regression, SVM only measures on which side of the hyperplane a sample lies and how far it is from that boundary. It does not inherently model class probabilities.

---

### Q4. What is the difference between the functional margin and the geometric margin?

The functional margin is the raw decision value scaled by the label and depends on the magnitude of \(w\). The geometric margin divides by \(||w||\), making it the true perpendicular distance from the hyperplane.
