# Ensemble Learning

An ensemble combines multiple models to produce one final prediction.

```
Model 1
Model 2
Model 3
...
Model N
    │
    ▼
Final Prediction
```

The idea is that **many weak or unstable models together often perform better than a single model**.

---

# Why Do We Need Bagging?

Suppose we train a Decision Tree.

```
Dataset
   │
   ▼
Decision Tree
   │
Accuracy = 91%
```

Now imagine changing just a few rows in the dataset.

```
Dataset A  → Tree A
Dataset B  → Tree B
Dataset C  → Tree C
```

The trees may become completely different.

This is because Decision Trees have **high variance**.

High variance means:

> Small changes in the training data cause large changes in the learned model.

Bagging is designed to reduce this variance.

---

# The Main Idea of Bagging

Instead of training one model,

train many models on different versions of the dataset.

```
Original Dataset
        │
        ├───────────┐
        │           │
Bootstrap 1   Bootstrap 2
        │           │
      Tree 1      Tree 2
        │           │
        └──────┬────┘
               ▼
        Majority Voting
```

Each tree makes mistakes in different places.

Voting usually removes many of those mistakes.

---

# Bootstrap

This is the most important concept.

Suppose the dataset contains only five samples.

```
1
2
3
4
5
```

Instead of using exactly these five samples,

Bagging creates a **new dataset** by randomly selecting samples **with replacement**.

Example Bootstrap Sample

```
2
5
1
2
4
```

Notice

```
2 appears twice

3 does not appear
```

This is completely normal.

Every bootstrap dataset has

* the same number of samples
* duplicate rows
* some missing rows

---

## Why "With Replacement"?

Imagine putting five balls in a box.

```
A
B
C
D
E
```

Pick one.

Suppose you pick

```
C
```

If you put it back,

```
A
B
C
D
E
```

it can be picked again.

That is called **sampling with replacement**.

---

Without replacement

```
Pick C

Remaining

A
B
D
E
```

C cannot appear again.

Bagging **always uses replacement**.

---

# Bootstrap Example

Original dataset

```
A
B
C
D
E
```

Tree 1

```
A
C
D
D
E
```

Tree 2

```
B
B
A
E
D
```

Tree 3

```
E
A
A
C
B
```

Notice

Every tree receives a different dataset.

# Bagging

Now every bootstrap dataset trains one model.

```
Bootstrap 1
      │
    Tree 1

Bootstrap 2
      │
    Tree 2

Bootstrap 3
      │
    Tree 3

...

Bootstrap 100
      │
    Tree 100
```

Finally,

```
Tree 1 → Spam

Tree 2 → Spam

Tree 3 → Ham

Tree 4 → Spam

Tree 5 → Spam
```

Majority vote

```
Spam
```

That becomes the final prediction.

# Pasting

Pasting is almost identical to Bagging.

The only difference is how the data is sampled.

Instead of

```
With replacement
```

it uses

```
Without replacement
```

Example

Original

```
A
B
C
D
E
```

Bagging

```
A
C
C
E
B
```

Duplicates allowed.

---

Pasting

```
A
C
E
B
D
```

No duplicates.

Every sample can appear only once.

---

## Comparison

| Bagging                   | Pasting                      |
| ------------------------- | ---------------------------- |
| Sampling with replacement | Sampling without replacement |
| Duplicate rows allowed    | No duplicate rows            |
| Some rows may disappear   | Each selected row is unique  |
| Most common               | Less common                  |

---

# Why Bagging Works Better Than Pasting

Imagine a dataset with 1,000 rows.

Bagging may produce

```
Row 25 appears four times.

Row 301 never appears.

Row 900 appears twice.
```

Every tree becomes more different from the others.

More diversity means

```
Less correlation
```

between trees.

Less correlated trees produce better ensemble performance.

Pasting creates datasets that are more similar to each other, so the models tend to be more correlated.

---

# Random Subspaces

Instead of changing the rows,

change the columns.

Example

Original dataset

```
Age
Salary
Height
Weight
Blood Pressure
Gender
Income
```

Tree 1

```
Age
Height
Income
```

Tree 2

```
Salary
Gender
Weight
```

Tree 3

```
Blood Pressure
Height
Income
```

Every model sees different features.

This is called the **Random Subspaces Method**.

It increases diversity when there are many features.

---

# Random Patches

Random Patches combine both ideas:

* Randomly select rows.
* Randomly select columns.

Example

Original

```
1000 rows

20 features
```

Tree 1

```
700 rows

8 features
```

Tree 2

```
700 rows

10 features
```

Tree 3

```
700 rows

6 features
```

Both the samples and the features are different for each model.

---

# Out-of-Bag (OOB) Samples

This is one of the biggest advantages of Bagging.

Suppose your dataset has five samples.

```
A
B
C
D
E
```

Bootstrap sample for one tree

```
A
A
B
D
E
```

Notice

```
C was never selected.
```

For this tree,

```
C
```

is an **Out-of-Bag (OOB)** sample.

The tree has **never seen** `C` during training.

So we can use `C` to test that tree.

Every tree leaves out different samples, and by combining these OOB predictions, we can estimate model performance **without creating a separate validation set**. On average, each bootstrap sample contains about **63.2% unique training samples**, leaving approximately **36.8%** as OOB samples.

---

# Summary

| Technique        | Rows   | Columns | Replacement               |
| ---------------- | ------ | ------- | ------------------------- |
| Bagging          | Random | All     | ✅ Yes                     |
| Pasting          | Random | All     | ❌ No                      |
| Random Subspaces | All    | Random  | Depends on implementation |
| Random Patches   | Random | Random  | Depends on implementation |

---

## Key Takeaways

* **Bagging** = Bootstrap sampling (with replacement) + Aggregating predictions.
* **Bootstrap** creates different training datasets by allowing duplicate samples.
* **Pasting** samples without replacement, so no duplicate rows appear.
* **Random Subspaces** create diversity by selecting different subsets of features.
* **Random Patches** create diversity by selecting both different rows and different features.
* **Out-of-Bag (OOB)** samples provide a built-in way to estimate generalization performance without needing a separate validation dataset.

These concepts form the foundation for understanding **Random Forest**, which is essentially **Bagging + Decision Trees + Random Feature Selection (Random Subspaces at each split)**. That will be the next major topic after mastering Bagging.
