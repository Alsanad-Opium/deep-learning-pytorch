```python
import optuna
from sklearn.datasets import  load_breast_cancer
from sklearn.linear_model import LogisticRegression, SGDClassifier
from sklearn.tree import DecisionTreeClassifier
import matplotlib.pyplot as plt 
import seaborn as sns
import pandas as pd 
from sklearn.preprocessing import StandardScaler 
from sklearn.model_selection import train_test_split,cross_val_score
from sklearn.metrics import classification_report,f1_score
from sklearn.pipeline import make_pipeline
```

```python
df  = load_breast_cancer(as_frame= True)
df = df.frame
df.head()
```

```python
df.describe()
```

```python
print(dict( df.skew() ))
```

```python
x_train,x_test, y_train, y_test = train_test_split(df.drop("target", axis=1), df.target, test_size=0.2, random_state=42)
```

```python
x_train.shape, x_test.shape
```

```python
sc = StandardScaler()
x_train_scaled  = sc.fit_transform(x_train)
x_test_scaled = sc.transform(x_test)
```

```python
x_test_scaled.shape
```

```python
model1 = LogisticRegression(max_iter=5000)
pipe =make_pipeline (StandardScaler(), model1)

model1.fit(x_train_scaled, y_train)
y_pred = model1.predict(x_test_scaled)

score = cross_val_score(pipe,x_train,y_train,scoring="f1",cv=3 )
print(score.mean())
print(classification_report(y_pred=y_pred,y_true= y_test))

```

```python
y_pred.shape, y_test.shape
```

```python
model2 = DecisionTreeClassifier(max_depth=7)
model2.fit(x_train,y_train)
y_pred = model2.predict(x_test)

score = cross_val_score(model2,x_train,y_train,scoring="f1",cv=3 )
print(score.mean())
print(classification_report(y_pred=y_pred, y_true= y_test))
```

```python
model3 = SGDClassifier()
model3.fit(x_train_scaled, y_train)
y_pred = model3.predict(x_test)


score = cross_val_score(model3,x_train,y_train,scoring="f1",cv=3 )
print(score.mean())
print(classification_report(y_pred,y_test))
```

```python
def objective_function(trial):
    model = trial.suggest_categorical('model_type' ,["LogisticRegression", "SDGClassifier", "DecisionTreeClassifer"])
    
    if model == "LogisticRegression":
        
        l1_ratio = trial.suggest_float('lr_l1_ratio', 0,1)
        solver  = trial.suggest_categorical('lr_solver', ["liblinear", "saga"])
        max_iter = trial.suggest_int('lr_max_iter', 200,5000)
        model = LogisticRegression(l1_ratio = l1_ratio, max_iter = max_iter)
        # model.fit(x_train_scaled,y_train)
        score = cross_val_score(model,x_train_scaled, y_train, cv = 5, scoring = "f1")
        
    elif model ==    "SDGClassifier":
        
        loss = trial.suggest_categorical('sdg_loss',["hinge", "modified_huber", "log_loss"])
        alpha = trial.suggest_float('sdg_alpha', 0.001,0.1)
        penalty = trial.suggest_categorical('sdg_penalty',["l1", "l2", "elasticnet", None])
        
        model = SGDClassifier(loss = loss, alpha = alpha, penalty = penalty)
        # model.fit(x_train_scaled, y_train)
        score = cross_val_score(model,x_train_scaled, y_train, cv = 5, scoring = "f1")
    
    else: 
        
        max_depth = trial.suggest_int('dt_max_depth',3,13)
        ccp_alpha = trial.suggest_int('dt_ccp', 0.1,1)
        model = DecisionTreeClassifier(max_depth = max_depth,ccp_alpha =  ccp_alpha, random_state = 42)  
        # model.fit(x_train,y_train)
        score = cross_val_score(model,x_train, y_train, cv = 5, scoring = "f1")
    
    return score.mean() 
     
        
        
```

```python
study =optuna.create_study(direction = "maximize")
study.optimize(objective_function,n_trials = 200)
```
