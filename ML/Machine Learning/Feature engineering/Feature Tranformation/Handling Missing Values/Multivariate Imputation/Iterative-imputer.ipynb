```python
import pandas as pd 
import numpy as np

df = pd.read_csv('Titanic-Dataset.csv')

df.isnull().sum()
```

```python
df['Age'].mean()
```

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute  import IterativeImputer

imp = IterativeImputer(max_iter=2, random_state=  42, n_nearest_features=2)

```

```python
imputed_age = imp.fit_transform(df[['Age']])
```

```python
print(imputed_age[727])
df['Age'].iloc[727]
```
