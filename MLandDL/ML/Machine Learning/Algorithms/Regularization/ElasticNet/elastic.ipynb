```python
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Lasso, Ridge, ElasticNet
from sklearn.metrics import mean_squared_error, r2_score
```

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("noopurbhatt/retail-black-friday-sales-dataset")

print("Path to dataset files:", path)
```

```python
df = pd.read_csv('retail_black_friday_sales_100k.csv')
df.head()
```

```python
df.columns
```

```python
df_new  = df.drop(['transaction_id', 'customer_id','payment_method', 'purchase_date', 'purchase_hour', 'is_weekend',
       'is_black_friday','gender', 'city','customer_segment', 'product_id', 'product_category','age_group'],axis=1)
df_new.columns
```

```python
x_train,x_test,y_train,y_test = train_test_split(df_new.drop('purchase_amount', axis=1), df['purchase_amount'], test_size=0.2, random_state=42)
x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

```python
lr = LinearRegression()
lr.fit(x_train, y_train)
lr_pred = lr.predict(x_test)
print("MSE :", mean_squared_error(y_test, lr_pred))
print("R2 Score :", r2_score(y_test, lr_pred))
```

```python
lasso = Lasso(alpha = 0.1)
lasso.fit(x_train, y_train)
lasso_pred = lasso.predict(x_test)
print("MSE:", mean_squared_error(y_test, lasso_pred))
print("R2 Score:", r2_score(y_test, lasso_pred))
```

```python
Ridge = Ridge(alpha = 0.5)
Ridge.fit(x_train, y_train)
Ridge_pred = Ridge.predict(x_test)
print("MSE:", mean_squared_error(y_test, Ridge_pred))
print("R2 Score:", r2_score(y_test, Ridge_pred))
```

```python
Elastic = ElasticNet(alpha = 0.01, l1_ratio = 0.01)
Elastic.fit(x_train, y_train)
Elastic_pred = Elastic.predict(x_test)
print("MSE:", mean_squared_error(y_test, Elastic_pred))
print("R2 Score:", r2_score(y_test, Elastic_pred))
```
