```python
import pandas as pd 
```

```python
df = pd.read_csv('./dataset/day.csv', index_col = 'dteday',parse_dates=['dteday'])

df.head()
```

```python
df = df.drop('instant', axis = 1)
df.head()
```

```python
print(list(df.columns))
print("Info :")
print(df.info())

print("Describe :")
print(df.describe())
```

```python
df.duplicated().sum()
```

```python
import seaborn as sns 
import matplotlib.pyplot as plt

plt.figure(figsize=(5,5))
sns.histplot(df.cnt, kde=True)
plt.title("Distribution of the data ")
plt.show()
```

```python
df.cnt.skew()
# The Target Variable is well distributed 
```

```python
sns.boxplot(data  = df, x = 'season', y = 'cnt')
plt.show()
```

```python
sns.boxplot(data = df , x ='cnt',color='red', showfliers= False)
sns.swarmplot(data=df ,x = 'cnt', size=3, alpha = 0.5)
plt.show()
```

```python
for col in df.columns:
    
    print('Column name:  {},  Values uniques: {}'.format(col,df[col].nunique() ))
```

```python
cat_col = ['season','yr','mnth', 'holiday', 'weekday', 'workingday', 'weathersit']

num_col  = ['temp', 'atemp', 'hum', 'windspeed', 'casual', 'registered', 'cnt']

for col in num_col:
    fig, ax = plt.subplots(1,2,figsize = (15,5))
    sns.histplot(df[col], kde = True, ax = ax[0])
    sns.boxplot(x = df[col], ax =ax[1])
    sns.swarmplot(x =df[col], alpha = 0.5,ax = ax[1])
    plt.suptitle(col)
    plt.show()
```

```python
## Inference 
# the windspeed and  casual , and temp are skewed

# windspeed : this means the windspeed is not the same everyday it varies very much, the data is slightly right skewed
# casual : This col is heavily right skewed which means most the number of non-registered users are less compared to some days 
# temp : this varies day to day it has somewhat of normal distribution  
```

```python
# Categorical variables

for col in cat_col:
    sns.countplot(y = df[col], order = df[col].value_counts().index)
    plt.title(col)
    plt.show()
```
