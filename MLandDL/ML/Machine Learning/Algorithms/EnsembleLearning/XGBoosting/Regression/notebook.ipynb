```python
for col in df.columns:
    
    print('Column name:  {},  Values uniques: {}'.format(col,df[col].nunique() ))
```

```python
cat_col = ['season','yr','mnth', 'holiday', 'weekday', 'workingday', 'weathersit']

num_col  = ['temp', 'atemp', 'hum', 'windspeed', 'casual', 'registered', 'cnt']

for col in num_col:
    fig, ax = plt.subplots(1,2,figsize = (15,5))
    sns.histplot(df[col], kde = True, ax = ax[0])
    sns.boxplot(x = df[col], ax =ax[1])
    sns.swarmplot(x =df[col], alpha = 0.5,ax = ax[1])
    plt.suptitle(col)
    plt.show()
```

```python
## Inference 
# the windspeed and  casual , and temp are skewed

# windspeed : this means the windspeed is not the same everyday it varies very much, the data is slightly right skewed
# casual : This col is heavily right skewed which means most the number of non-registered users are less compared to some days 
# temp : this varies day to day it has somewhat of normal distribution  
```

```python
# Categorical variables

for col in cat_col:
    sns.countplot(y = df[col], order = df[col].value_counts().index)
    plt.title(col)
    plt.show()
```

```python
# BIvariate analysis 

for col in num_col:
    
    if col != 'cnt':
        sns.regplot(x = df[col], y = df['cnt'],scatter_kws={"alpha":0.5} )
        plt.title(f"{col} vs CNT")
        plt.show()
            
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error,r2_score,root_mean_squared_error

df_after_drop = df.drop(['cnt','registered','casual','atemp'],axis = 1)
x_train,x_test,y_train,y_test = train_test_split(df.drop(['cnt','registered','casual','atemp'],axis = 1),df['cnt'],test_size=0.2, random_state=42)
```

```python
xgboost = xg.XGBRegressor(random_state = 42)
xgboost.fit(x_train,y_train)
y_pred = xgboost.predict(x_test)
print(mean_squared_error(y_pred,y_test))
print(root_mean_squared_error(y_pred,y_test))
print(r2_score(y_pred,y_test))
```
