```python
import pandas as pd 
```

```python
df = pd.read_csv('./dataset/day.csv', index_col = 'dteday',parse_dates=['dteday'])

df.head()
```

```python
df = df.drop('instant', axis = 1)
df.head()
```

```python
print(list(df.columns))
print("Info :")
print(df.info())

print("Describe :")
print(df.describe())
```

```python
import seaborn as sns 
import matplotlib.pyplot as plt

plt.figure(figsize=(5,5))
sns.histplot(df.cnt, kde=True)
plt.title("Distribution of the data ")
plt.show()
```

```python
df.cnt.skew()
# The Target Variable is well distributed 
```

```python
sns.boxplot(data  = df, x = 'season', y = 'cnt')
plt.show()
```

```python
sns.boxplot(data = df , x ='cnt',color='red', showfliers= False)
sns.swarmplot(data=df ,x = 'cnt', size=3, alpha = 0.5)
plt.show()
```
