```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
from sklearn.ensemble import VotingClassifier
```

```python
df = pd.read_csv('heart.csv')
df.head()
```

```python
df.info()
```

```python
df.duplicated().sum()
df = df.drop_duplicates()
df.duplicated().sum()
```

```python
new_df  = pd.get_dummies(df, columns=['cp','sex','fbs', 'restecg','exang','slope','ca', 'thal'  ],dtype= float )
```

```python
new_df.head()
```

```python
x_new = new_df.drop('target', axis =1)
y_new = new_df['target']
x_new.shape,y_new.shape
```

```python
sc = StandardScaler()

x_new = sc.fit_transform(x_new)
```

```python
x_new.shape
```

```python
x = df.drop('target',axis =1)
y =df['target']
x.shape,y.shape
```

```python
from sklearn.model_selection import cross_val_score
import numpy as np 

best_k = []
for value in range(1,31):
    knn = KNeighborsClassifier(n_neighbors=value)
    score = cross_val_score(knn,x_new,y_new,cv = 10, scoring='accuracy')
    best_k.append(score.mean())

best_k_arr = np.array(best_k)
print(best_k_arr.argmax()+1, best_k_arr.max())
print(best_k)
best_k_val = best_k_arr.argmax()+1
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(x_new,y, test_size =0.2,random_state = 42)
```

```python
lor = LogisticRegression(max_iter=5000)
gnb = GaussianNB(var_smoothing=0.0001)
svc = SVC(probability=True)
knn =KNeighborsClassifier(n_neighbors=best_k_val)


estimators = [('lr',lor), ('gnb', gnb), ('svc', svc), ('knn', knn)]
```

```python
for name,model in estimators:
    score = []
    score = cross_val_score(model, x_new,y_new,cv = 10, scoring='accuracy')
    print(name,score.mean())
    # model.fit(x_train,y_train)
    # y_pred = model.predict(x_test)
    # print(accuracy_score(y_test,y_pred), name)
```

```python
voting = VotingClassifier(estimators=estimators, voting='soft')
score = cross_val_score(voting,x_new,y_new,cv =10, scoring='accuracy')
print(score.mean())
```

```python
voting = VotingClassifier(estimators=estimators, voting='hard')
score = cross_val_score(voting,x_new,y_new,cv =10, scoring='accuracy')
print(score.mean())
```
