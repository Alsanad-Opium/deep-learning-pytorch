# 📘 Logistic Regression — Complete Notes

---

# 🧠 1. What is Logistic Regression?

Logistic Regression is a **supervised machine learning algorithm** used for **binary classification** problems.

* Output → Probability (0 to 1)
* Decision → Class label (0 or 1)

---

# 📌 2. Model Equation

## Linear Combination

[
z = w_1x_1 + w_2x_2 + \dots + w_nx_n + b
]

---

## Sigmoid Function (Activation)

[
\sigma(z) = \frac{1}{1 + e^{-z}}
]

* Converts any real value → (0, 1)
* Interpreted as probability

---

# 🔥 3. Prediction Rule

[
\hat{y} =
\begin{cases}
1 & \text{if } p \geq 0.5 \
0 & \text{if } p < 0.5
\end{cases}
]

---

# 📉 4. Loss Function (Log Loss / Binary Cross Entropy)

[
L = -\frac{1}{n} \sum \left[ y \log(p) + (1 - y)\log(1 - p) \right]
]

### Why not MSE?

* Non-convex for classification
* Poor gradients

---

# ⚙️ 5. Optimization (Gradient Descent)

Goal: Minimize loss

---

## Gradients

[
\frac{\partial L}{\partial w} = \frac{1}{n} X^T (p - y)
]

[
\frac{\partial L}{\partial b} = \frac{1}{n} \sum (p - y)
]

---

## Update Rule

[
w := w - \alpha \cdot dw
]

[
b := b - \alpha \cdot db
]

---

# 🚀 6. Training Algorithm

1. Initialize weights and bias
2. Compute ( z = wx + b )
3. Apply sigmoid → ( p )
4. Compute loss
5. Compute gradients
6. Update weights
7. Repeat until convergence

---

# 💻 7. From Scratch Implementation

```python
import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def compute_loss(y, p):
    p = np.clip(p, 1e-15, 1 - 1e-15)
    return -np.mean(y*np.log(p) + (1-y)*np.log(1-p))

def train(X, y, lr=0.01, epochs=1000):
    n, m = X.shape
    w = np.zeros(m)
    b = 0

    for _ in range(epochs):
        z = np.dot(X, w) + b
        p = sigmoid(z)

        dw = np.dot(X.T, (p - y)) / n
        db = np.mean(p - y)

        w -= lr * dw
        b -= lr * db

    return w, b
```

# 🔍 8. Decision Boundary

[
w_1x_1 + w_2x_2 + b = 0
]

* Linear separator
* Works well for linearly separable data

---

# ⚠️ 9. Assumptions

* Linear relationship (log-odds)
* No multicollinearity
* Independent observations

---

# 📊 10. Evaluation Metrics

* Accuracy
* Precision
* Recall
* F1 Score
* ROC-AUC

---

# 🔧 11. Regularization

## L2 (Ridge)

[
L + \lambda \sum w^2
]

* Shrinks weights
* Prevents overfitting

---

## L1 (Lasso)

[
L + \lambda \sum |w|
]

* Forces some weights to 0
* Feature selection

---

# ⚡ 12. Odds and Log-Odds

[
\text{odds} = \frac{p}{1-p}
]

[
\log(\text{odds}) = wx + b
]

---

# 🧩 13. Interpretation of Coefficients

* Positive → increases probability
* Negative → decreases probability
* Magnitude → strength (after scaling)

---

# 🚨 14. Common Mistakes

* ❌ Using label encoding for nominal data
* ❌ Not scaling features
* ❌ Interpreting coefficients without scaling
* ❌ Using MSE loss
