# 🧠 1. What is Binary Cross Entropy?

Binary Cross Entropy (BCE) = **loss function for binary classification**

👉 It measures how far your predicted probability is from the actual label.

---

# 📌 Formula

[
L = -\frac{1}{n} \sum \left[ y \log(p) + (1 - y)\log(1 - p) \right]
]

Where:

* ( y \in {0,1} ) → actual label
* ( p ) → predicted probability

---

# 🔥 2. Intuition (VERY IMPORTANT)

## Case 1: Correct Prediction

| y | p    | Loss  |
| - | ---- | ----- |
| 1 | 0.99 | LOW ✅ |
| 0 | 0.01 | LOW ✅ |

👉 Model is confident and correct → small loss

---

## Case 2: Wrong Prediction

| y | p    | Loss   |
| - | ---- | ------ |
| 1 | 0.01 | HIGH ❌ |
| 0 | 0.99 | HIGH ❌ |

👉 Model is confident and wrong → HUGE penalty

---

# ⚙️ 3. Simplified View

## If y = 1:

[
L = -\log(p)
]

## If y = 0:

[
L = -\log(1 - p)
]

---

# 📈 4. Why This Works

* Penalizes **wrong confident predictions heavily**
* Smooth + convex → easy to optimize
* Works perfectly with sigmoid output

---

# 🔗 5. Connection to Logistic Regression

Pipeline:

```text
X → z = wx + b → sigmoid → p → BCE loss
```

👉 BCE is applied **after sigmoid**

---

# 💻 6. Code (Manual Implementation)

```python
import numpy as np

def binary_cross_entropy(y, p):
    p = np.clip(p, 1e-15, 1 - 1e-15)  # avoid log(0)
    return -np.mean(y*np.log(p) + (1-y)*np.log(1-p))

# example
y = np.array([1, 0, 1, 0])
p = np.array([0.9, 0.2, 0.8, 0.1])

loss = binary_cross_entropy(y, p)
print(loss)
```
