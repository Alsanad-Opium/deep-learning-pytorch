# Decision Trees - Part 5.1: Building the Tree (Learning Algorithm & Recursive Partitioning)

---

# Introduction

So far, we have learned:

- Tree structure
- Pure and impure nodes
- Entropy
- Information Gain
- Gini Impurity

Now the next question is:

> **How does the Decision Tree use these concepts to build an entire tree?**

A Decision Tree does not create the whole tree at once.

Instead, it builds the tree **one node at a time**, starting from the root and repeatedly splitting the data until a stopping condition is reached.

This process is known as **Recursive Partitioning**.

---

# High-Level Overview

The Decision Tree follows the same sequence repeatedly.

```text
Entire Dataset

↓

Find Best Split

↓

Create Child Nodes

↓

Repeat on Left Child

↓

Repeat on Right Child

↓

Stop When Necessary

↓

Final Decision Tree
```

Notice that the algorithm performs the same task again and again.

This repeated process is called **recursion**.

---

# What is Recursive Partitioning?

## Definition

Recursive Partitioning is the process of repeatedly dividing a dataset into smaller subsets until each subset becomes sufficiently pure or satisfies a stopping criterion.

The word **partition** means:

> Divide something into smaller parts.

The word **recursive** means:

> Perform the same operation repeatedly.

Therefore,

Recursive Partitioning simply means:

> Keep splitting the data into smaller and smaller groups.

---

# Why Recursion?

Imagine organizing books in a library.

Initially,

everything is mixed together.

```text
Books

↓

Science
History
Math
Programming
Biology
```

First,

separate by subject.

Now,

inside Programming,

separate by language.

```text
Programming

↓

Python

Java

C++
```

Inside Python,

separate again.

```text
Python

↓

Beginner

Intermediate

Advanced
```

The same process repeats at every level.

Decision Trees work in exactly the same way.

---

# The Root Node

Every Decision Tree starts with a single node.

```text
Entire Dataset
```

Example

| Age | Student | Buy Laptop |
|------|----------|------------|
| Young | Yes | Yes |
| Young | No | No |
| Old | Yes | Yes |
| Old | No | No |
| Young | Yes | Yes |
| Old | No | No |

Initially,

every training sample belongs to the root node.

---

# Step 1: Calculate Node Impurity

The algorithm first measures the impurity of the root.

Depending on the criterion,

it calculates either:

- Entropy

or

- Gini Impurity

Example

```text
Root Node

↓

Entropy = 0.97
```

---

# Step 2: Try Every Feature

Suppose our dataset contains

```text
Age

Salary

Student

Experience
```

The algorithm does **not** immediately choose one feature.

Instead,

it evaluates every feature.

```text
Age ?

↓

Information Gain = ?

Salary ?

↓

Information Gain = ?

Student ?

↓

Information Gain = ?

Experience ?

↓

Information Gain = ?
```

---

# Step 3: Try Every Possible Split

For numerical features,

the algorithm also tries different thresholds.

Example

Age

```text
Age < 20

Age < 25

Age < 30

Age < 35
```

Each threshold produces a different split.

Each split has a different impurity.

---

# Example

Suppose

```text
Age

18

20

24

30

35

40
```

Possible thresholds

```text
Age < 19

Age < 22

Age < 27

Age < 33

Age < 38
```

The algorithm evaluates every candidate.

---

# Step 4: Choose the Best Split

After computing Information Gain (or Gini Decrease) for every candidate,

the algorithm chooses

```text
Maximum Information Gain
```

or

```text
Minimum Weighted Gini
```

Example

| Split | Information Gain |
|--------|-----------------:|
| Age < 22 | 0.15 |
| Age < 30 | 0.48 |
| Salary < 50K | 0.27 |
| Student = Yes | 0.64 |

Best split

```text
Student = Yes
```

---

# Step 5: Create Child Nodes

Suppose the chosen split is

```text
Student?
```

The dataset is divided.

```text
Student?

       Yes             No

     Child A       Child B
```

Now,

each child contains a subset of the original data.

---

# Step 6: Repeat the Same Process

The algorithm now treats each child as a completely new dataset.

For Child A,

```text
Calculate Impurity

↓

Try Every Feature

↓

Find Best Split

↓

Split Again
```

Exactly the same process happens for Child B.

---

# Recursive Nature

Notice what happens.

```text
Root

↓

Split

↓

Left Child

↓

Split Again

↓

Left Child

↓

Split Again

↓

...
```

The algorithm repeatedly calls itself.

This is why the learning algorithm is recursive.

---

# Visualization

```text
Entire Dataset

↓

Best Split

↓

───────────────

│             │

Left         Right

│             │

Split       Split

│             │

More        More

Splits      Splits

│             │

Leaves      Leaves
```

---

# Why Is It Called a Greedy Algorithm?

Decision Trees make the **best decision at the current node**.

They do **not** check whether another split might produce a better overall tree later.

Example

Suppose two splits exist.

Split A

Immediate Information Gain

```text
0.65
```

Split B

Immediate Information Gain

```text
0.60
```

Even if Split B eventually produces a better overall tree,

the algorithm still chooses Split A because

```text
0.65 > 0.60
```

This behavior is called a **Greedy Algorithm**.

---

# Why Don't We Search Every Possible Tree?

Imagine only four features.

Even then,

there are thousands of possible tree structures.

For ten or twenty features,

the number becomes astronomically large.

Searching every possible tree would be computationally impossible.

Therefore,

Decision Trees use the greedy strategy,

which is much faster.

---

# Binary Splitting

Most modern Decision Tree implementations use **Binary Splits**.

Each node creates only two children.

Example

```text
Age < 30 ?

      Yes

      No
```

Not

```text
Young

Middle

Old
```

Binary trees are easier to optimize and implement.

The CART algorithm always uses binary splits.

---

# How Continuous Features Are Split

Suppose

```text
Salary

20K

25K

30K

35K

40K

50K
```

The algorithm considers thresholds between adjacent values.

```text
22.5K

27.5K

32.5K

37.5K

45K
```

Each threshold is evaluated.

The one producing the best impurity reduction is selected.

---

# What About Categorical Features?

Example

```text
Color

Red

Blue

Green
```

The algorithm creates splits such as

```text
Color = Red ?

Yes

No
```

or

```text
Color ∈ {Red, Blue} ?

Yes

No
```

The exact handling depends on the Decision Tree implementation.

---

# General Learning Algorithm

The complete learning process can be summarized as:

```text
Start

↓

Create Root Node

↓

Measure Impurity

↓

For Every Feature

↓

For Every Possible Split

↓

Calculate Information Gain
(or Gini)

↓

Choose Best Split

↓

Create Child Nodes

↓

Repeat Recursively

↓

Stopping Condition?

↓

No

↓

Split Again

↓

Yes

↓

Create Leaf Node
```

---

# Pseudocode

```text
BuildTree(Node)

Calculate impurity

If stopping criterion reached

    Make leaf

Else

    Find best split

    Divide data

    BuildTree(left child)

    BuildTree(right child)
```

Notice how the function calls itself.

This is recursion.

---

# Time Complexity (High-Level)

Training a Decision Tree involves:

- Trying many features
- Trying many thresholds
- Computing impurity repeatedly

Therefore,

training is more computationally expensive than prediction.

Prediction is very fast because the model simply follows one path from the root to a leaf.

We'll study the exact time and space complexity in a later chapter.

---

# Important Interview Questions

### Why is the Decision Tree learning algorithm recursive?

Because the same splitting procedure is applied repeatedly to every newly created child node.

---

### Why is the algorithm called greedy?

Because it selects the locally best split at each node without searching for the globally optimal tree.

---

### Why doesn't a Decision Tree search every possible tree?

The number of possible trees grows exponentially, making exhaustive search computationally infeasible.

---

### What is recursive partitioning?

It is the repeated process of dividing the dataset into smaller subsets until a stopping criterion is met.

---

### Does the algorithm split every feature?

Yes.

At each node, every candidate feature (and, for numerical features, many candidate thresholds) is evaluated before selecting the best split.

---

# Key Takeaways

- A Decision Tree is built **top-down**, beginning with the root node.
- At each node, the algorithm evaluates every possible split using an impurity measure such as Entropy or Gini.
- The best split is chosen greedily based on maximum Information Gain or minimum weighted Gini.
- After splitting, each child node is treated as a new dataset, and the process repeats recursively.
- Modern implementations such as CART use **binary splits**, creating exactly two child nodes at each decision node.
- The recursive process continues until a stopping criterion is reached, producing the final Decision Tree.

# Decision Trees - Part 5.2: Stopping Criteria, Tree Complexity, Overfitting, and Underfitting

---

# Introduction

In the previous chapter, we learned how a Decision Tree is built.

The algorithm repeatedly:

- Finds the best split
- Creates child nodes
- Recursively splits the children
- Continues until a stopping condition is met

This naturally raises an important question:

> **When should the Decision Tree stop growing?**

If the tree stops too early, it may fail to learn important patterns.

If it grows too much, it may memorize the training data.

Finding the right balance is one of the most important aspects of building a Decision Tree.

---

# Why Can't We Keep Splitting Forever?

Imagine a dataset with 100 training samples.

If we continue splitting until every leaf contains exactly **one sample**, the tree will classify every training example correctly.

Example:

```text
Leaf 1

Spam
```

```text
Leaf 2

Not Spam
```

```text
Leaf 3

Spam
```

...

Every leaf becomes perfectly pure.

Training Accuracy

```text
100%
```

At first glance, this seems perfect.

However,

the model has simply memorized the training data.

It has not learned general patterns.

---

# Generalization

The purpose of Machine Learning is **not** to memorize the training data.

The goal is to perform well on **new, unseen data**.

This ability is called **Generalization**.

A good Decision Tree should:

- Learn meaningful patterns
- Ignore random noise
- Make accurate predictions on unseen data

---

# Tree Complexity

As a Decision Tree grows,

its complexity increases.

Complexity depends on factors such as:

- Tree depth
- Number of nodes
- Number of leaves
- Number of splits

Example

### Simple Tree

```text
        Age?

      /      \

   Buy      Don't
```

Depth

```text
1
```

---

### Medium Tree

```text
           Age?

      /             \

 Student?        Salary?

  /     \          /    \

Buy   Don't     Buy   Don't
```

Depth

```text
2
```

---

### Complex Tree

```text
Age

↓

Student

↓

Salary

↓

Experience

↓

City

↓

Income

↓

Education

↓

...
```

Depth

```text
15+
```

The deeper the tree,

the more complex the model becomes.

---

# Underfitting

## Definition

**Underfitting** occurs when the Decision Tree is **too simple** to capture the underlying patterns in the data.

It performs poorly on both the training and testing datasets.

---

# Example

Suppose the true decision depends on:

- Age
- Income
- Education

But the tree only asks:

```text
Age > 30?
```

Prediction is made immediately.

The model ignores important information.

---

# Characteristics of Underfitting

- Very shallow tree
- High bias
- Low variance
- Poor training accuracy
- Poor testing accuracy

---

# Visualization

```text
Simple Tree

          Age?

       /       \

     Buy     Don't
```

Training Accuracy

```text
70%
```

Testing Accuracy

```text
68%
```

The model is too simple.

---

# Reasons for Underfitting

- Maximum depth too small
- Too few nodes
- Early stopping
- Very large minimum leaf size
- Excessive pruning

---

# Overfitting

## Definition

**Overfitting** occurs when the Decision Tree becomes **too complex**.

Instead of learning general patterns,

it memorizes the training data,

including random noise.

---

# Example

Imagine a dataset with one noisy observation.

Instead of ignoring it,

the tree creates several additional splits just to classify that single example correctly.

Example

```text
Age?

↓

Salary?

↓

Education?

↓

Experience?

↓

City?

↓

ZIP Code?

↓

Favorite Color?

↓

Prediction
```

Most of these splits are unnecessary.

---

# Characteristics of Overfitting

- Very deep tree
- Low bias
- High variance
- Extremely high training accuracy
- Lower testing accuracy

---

# Visualization

```text
Huge Tree

↓

Many Levels

↓

Hundreds of Nodes

↓

Tiny Leaves

↓

Memorized Data
```

Training Accuracy

```text
100%
```

Testing Accuracy

```text
78%
```

---

# Comparing Underfitting and Overfitting

| Property | Underfitting | Good Fit | Overfitting |
|----------|--------------|----------|-------------|
| Tree Depth | Too Small | Balanced | Too Large |
| Training Accuracy | Low | High | Very High |
| Testing Accuracy | Low | High | Lower |
| Bias | High | Balanced | Low |
| Variance | Low | Balanced | High |
| Generalization | Poor | Excellent | Poor |

---

# Bias-Variance Tradeoff

One of the most important concepts in Machine Learning.

---

## High Bias

Model is too simple.

Cannot learn the actual relationship.

Example

```text
Depth = 1
```

Prediction errors remain high,

even on the training data.

---

## High Variance

Model is too complex.

Small changes in training data produce very different trees.

Example

```text
Depth = 25
```

The model memorizes the data.

---

## Balanced Model

```text
Depth = 5
```

Captures useful patterns,

ignores noise,

generalizes well.

---

# Bias-Variance Visualization

```text
Prediction Error

^

|

|\
| \
|  \
|   \
|    \__
|        \__
|           \_____

+------------------------------>

Model Complexity
```

As complexity increases:

- Bias decreases
- Variance increases

The best model lies somewhere in the middle.

---

As the tree grows,

training accuracy almost always increases.

Testing accuracy,

Example

| Tree Depth | Train Accuracy | Test Accuracy |

# Why Do Decision Trees Overfit Easily?

Decision Trees have **high flexibility**.

They can create many small regions in the feature space.

Example

```text
Feature Space

↓

Split

↓

Split Again

↓

Split Again

↓

Tiny Regions
```

Eventually,

some leaves may contain only one or two samples.

The tree begins memorizing noise rather than learning general rules.

---

# Indicators of Overfitting

- Very deep tree
- Hundreds of nodes
- Many leaves with one sample
- Training accuracy near 100%
- Large gap between train and test accuracy

---

# Indicators of Underfitting

- Very shallow tree
- Low training accuracy
- Low testing accuracy
- Misses obvious relationships
- Poor predictions on both datasets

# How Can We Prevent Overfitting?

Several techniques help control tree growth.

Examples:

- Maximum depth
- Minimum samples split

These techniques will be covered in the next chapter.

# Stopping Criteria

A Decision Tree stops growing when **any** stopping condition is satisfied.

Common stopping criteria include:

---

## 1. Pure Node

If every sample belongs to the same class,

stop.

Example

```text
Spam

Spam

Spam
```

No further split is needed.

---

## 2. Maximum Depth Reached

Suppose

```text
max_depth = 5
```

Once depth 5 is reached,

the algorithm stops creating child nodes,

even if impurity still exists.

---

## 3. Minimum Samples Required to Split

Suppose

```text
min_samples_split = 10
```

If a node contains fewer than 10 samples,

it cannot be split further.

---

## 4. Minimum Samples per Leaf

Suppose

```text
min_samples_leaf = 5
```

Every leaf must contain at least 5 samples.

If a split would create a leaf with only 3 samples,

that split is rejected.

---

## 5. Maximum Number of Leaf Nodes

Example

```text
max_leaf_nodes = 20
```

Once 20 leaves have been created,

tree growth stops.

---

## 6. No Improvement in Impurity

If every possible split produces negligible improvement,

the algorithm stops.

There is no benefit in further partitioning the data.

---

# Complete Stopping Flow

```text
Current Node

↓

Pure?

↓

Yes → Stop

↓

No

↓

Max Depth Reached?

↓

Yes → Stop

↓

No

↓

Enough Samples?

↓

No → Stop

↓

Yes

↓

Improvement Exists?

↓

No → Stop

↓

Yes

↓

Split Again
```

### Why do Decision Trees overfit easily?

Because they continue splitting until they memorize the training data, including random noise.

---

### Why is training accuracy usually higher than testing accuracy?

---

### What is underfitting?

When the tree is too simple to capture the underlying patterns in the data.

---

### What is the bias-variance tradeoff?

Increasing model complexity reduces bias but increases variance. A good model balances both.

---

### Name common stopping criteria.

- Pure node

# Key Takeaways

- Decision Trees must stop growing to avoid memorizing the training data.
- **Underfitting** occurs when the tree is too simple, leading to poor performance on both training and testing data.
- **Overfitting** occurs when the tree becomes too complex and learns noise from the training data.
- The bias-variance tradeoff explains why neither an overly simple nor an overly complex tree performs best.
- Stopping criteria such as `max_depth`, `min_samples_split`, and `min_samples_leaf` help control tree complexity.
- A well-balanced tree achieves strong training performance while maintaining good generalization on unseen data.

# Decision Trees - Part 5.3: Pre-Pruning, Post-Pruning, Cost Complexity Pruning, and Hyperparameters

---

# Introduction

In the previous chapter, we learned that Decision Trees tend to **overfit** when allowed to grow without restriction.

A very deep tree:

- Memorizes training data
- Learns random noise
- Performs poorly on unseen data

Therefore, we need techniques to **control tree growth**.

There are two major approaches:

1. **Pre-Pruning (Early Stopping)** – Stop the tree from growing too much.
2. **Post-Pruning** – Grow the full tree first, then remove unnecessary branches.

Both techniques aim to improve **generalization**.

---

# What is Pruning?

## Definition

**Pruning** is the process of reducing the size of a Decision Tree by limiting or removing unnecessary branches.

The objective is to:

- Reduce overfitting
- Improve generalization
- Make the tree simpler
- Reduce prediction time
- Improve interpretability

---

# Why is Pruning Needed?

Suppose we build a tree without restrictions.

```text
                    Root

                 /        \

              Node        Node

            /    \       /     \

         Node   Node   Node   Node

        /  \   /  \    /  \   /  \

      ... many more levels ...
```

Eventually,

some leaves may contain only **one training sample**.

Although training accuracy becomes very high,

the tree performs poorly on new data.

Pruning prevents this problem.

---

# Types of Pruning

There are two major approaches.

```text
Pruning

│

├── Pre-Pruning

│

└── Post-Pruning
```

---

# Pre-Pruning (Early Stopping)

## Definition

**Pre-Pruning** stops the tree **before** it becomes too large.

Instead of allowing unlimited growth,

the algorithm follows predefined stopping rules.

---

# Intuition

Imagine building a road.

Instead of building unnecessary extra roads,

you stop once the required destinations are connected.

Similarly,

a Decision Tree stops growing once it has learned enough useful patterns.

---

# Common Pre-Pruning Hyperparameters

- `max_depth`
- `min_samples_split`
- `min_samples_leaf`
- `max_leaf_nodes`
- `max_features`
- `min_impurity_decrease`

These hyperparameters control tree growth.

---

# 1. max_depth

## Definition

Maximum number of levels allowed in the tree.

Example

```python
DecisionTreeClassifier(max_depth=3)
```

Example tree

```text
Depth 0

Root

↓

Depth 1

↓

Depth 2

↓

Depth 3

↓

Stop
```

No node can grow beyond depth 3.

---

## Effect

Small value

```text
Simple Tree

↓

Higher Bias

↓

Lower Variance
```

Large value

```text
Deep Tree

↓

Lower Bias

↓

Higher Variance
```

---

# Advantages

- Prevents overfitting
- Faster training
- Faster prediction
- Easier visualization

---

# Disadvantages

If chosen too small,

important patterns may be missed.

---

# 2. min_samples_split

## Definition

Minimum number of samples required before a node can be split.

Example

```python
DecisionTreeClassifier(min_samples_split=10)
```

Suppose

```text
Current Node

8 Samples
```

Since

```text
8 < 10
```

the node cannot split.

---

## Why Useful?

Prevents splitting very small groups,

which often represent noise.

---

# 3. min_samples_leaf

## Definition

Minimum number of samples required in every leaf node.

Example

```python
DecisionTreeClassifier(min_samples_leaf=5)
```

Suppose a split produces

```text
Left Leaf

2 Samples

Right Leaf

18 Samples
```

Since

```text
2 < 5
```

this split is rejected.

---

## Why Important?

Prevents tiny leaves,

which often overfit.

---

# Difference Between min_samples_split and min_samples_leaf

| min_samples_split | min_samples_leaf |
|-------------------|------------------|
| Checks the parent node before splitting | Checks the child nodes after splitting |
| Determines whether a split is allowed | Ensures each resulting leaf has enough samples |

---

# 4. max_leaf_nodes

Limits the maximum number of leaves.

Example

```python
DecisionTreeClassifier(max_leaf_nodes=20)
```

After creating 20 leaves,

the tree stops growing.

---

# Why Useful?

Limits overall tree complexity.

---

# 5. max_features

Instead of considering every feature,

the algorithm examines only a subset.

Example

```python
DecisionTreeClassifier(max_features="sqrt")
```

Useful for:

- Faster training
- Reduced correlation in ensemble methods (e.g., Random Forest)

---

# 6. min_impurity_decrease

A split is accepted only if it decreases impurity by at least a specified amount.

Example

```python
DecisionTreeClassifier(min_impurity_decrease=0.01)
```

Tiny improvements are ignored.

---

# Summary of Pre-Pruning Parameters

| Hyperparameter | Purpose |
|----------------|---------|
| `max_depth` | Limits tree depth |
| `min_samples_split` | Minimum samples before splitting |
| `min_samples_leaf` | Minimum samples in each leaf |
| `max_leaf_nodes` | Maximum number of leaves |
| `max_features` | Limits candidate features |
| `min_impurity_decrease` | Requires meaningful impurity reduction |

---

# Advantages of Pre-Pruning

- Prevents overfitting
- Faster training
- Lower memory usage
- Smaller trees
- Easier interpretation

---

# Disadvantages of Pre-Pruning

- May stop too early
- Can underfit
- Requires hyperparameter tuning

---

# Post-Pruning

## Definition

Instead of stopping early,

grow the **complete tree first**,

then remove branches that contribute little to predictive performance.

---

# Intuition

Imagine sculpting a statue.

You first carve a rough block,

then remove unnecessary material.

Decision Trees follow the same idea.

---

# Workflow

```text
Grow Full Tree

↓

Evaluate Branches

↓

Remove Weak Branches

↓

Simpler Tree

↓

Better Generalization
```

---

# Why Post-Pruning Often Performs Better

Growing the full tree allows the algorithm to discover all possible patterns.

Pruning later removes only those branches that do not improve generalization.

This often produces better results than stopping too early.

---

# Cost Complexity Pruning (CCP)

The most common post-pruning technique used by **scikit-learn**.

---

## Idea

Each branch has:

- A benefit (reduced impurity)
- A cost (increased complexity)

Cost Complexity Pruning removes branches whose cost outweighs their benefit.

---

# Cost Complexity Formula

Conceptually,

the algorithm minimizes

```text
Total Cost

=

Classification Error

+

α × Tree Complexity
```

where

- Classification Error measures prediction mistakes.
- Tree Complexity is related to the number of leaf nodes.
- α (alpha) controls the trade-off.

A larger α penalizes large trees more heavily.

---

# ccp_alpha

The parameter controlling Cost Complexity Pruning.

Example

```python
DecisionTreeClassifier(ccp_alpha=0.01)
```

---

## Interpretation

### Small α

```text
Minimal Pruning

↓

Large Tree
```

---

### Large α

```text
Aggressive Pruning

↓

Small Tree
```

---

# Choosing the Best ccp_alpha

Instead of guessing,

we usually:

1. Train multiple trees using different α values.
2. Evaluate them using cross-validation.
3. Select the α with the best validation performance.

---

# Visualization

```text
Full Tree

↓

ccp_alpha = 0

↓

Huge Tree

↓

Increase Alpha

↓

Remove Branches

↓

Smaller Tree

↓

Better Generalization
```

# Choosing Hyperparameters

Common approach:

```text
Split Data

↓

Training Set

↓

Cross Validation

↓

Try Many Hyperparameters

↓

Choose Best Model

↓

Evaluate on Test Set
```

Grid Search and Random Search are often used to automate this process.

---

# Pre-Pruning vs Post-Pruning

| Feature | Pre-Pruning | Post-Pruning |
|----------|-------------|--------------|
| When applied | During tree growth | After full tree is built |
| Tree size | Smaller from the start | Initially very large |
| Training time | Usually faster | Usually slower |
| Risk | Underfitting | Lower risk of underfitting |
| Flexibility | Lower | Higher |

---

# Which One Should You Use?

In practice,

both are often combined.

Example:

```python
DecisionTreeClassifier(
    max_depth=8,
    min_samples_leaf=5,
    ccp_alpha=0.002
)
```

The tree:

- Stops excessive growth.
- Removes unnecessary branches afterward.

This often provides the best balance between bias and variance.

---

# Important Interview Questions

### What is pruning?

Pruning is the process of reducing the size of a Decision Tree to improve generalization and reduce overfitting.

---

### What is the difference between pre-pruning and post-pruning?

Pre-pruning stops tree growth early, whereas post-pruning first grows the full tree and then removes unnecessary branches.

---

### What does `max_depth` control?

The maximum depth (number of levels) of the Decision Tree.

---

### What is the difference between `min_samples_split` and `min_samples_leaf`?

`min_samples_split` controls whether a node can split, while `min_samples_leaf` ensures that each resulting leaf contains a minimum number of samples.

---

### What is `ccp_alpha`?

`ccp_alpha` controls Cost Complexity Pruning. Larger values produce smaller, more aggressively pruned trees.

---

# Key Takeaways

- Pruning reduces overfitting by controlling tree complexity.
- **Pre-Pruning** limits tree growth during training using hyperparameters such as `max_depth` and `min_samples_leaf`.
- **Post-Pruning** grows the full tree first and then removes unnecessary branches.
- Cost Complexity Pruning balances prediction error against tree complexity using the `ccp_alpha` parameter.
- Hyperparameter tuning is essential for achieving a Decision Tree that generalizes well to unseen data.
- In practice, combining pre-pruning with post-pruning often yields the best results.

# What is a Regression Tree?

## Definition

A **Regression Tree** is a Decision Tree that predicts **continuous numerical values** instead of class labels.

Instead of asking:

```text
Spam?

Not Spam?
```

it predicts values such as

```text
₹45,20,000
```

or

```text
Temperature = 31.5°C
```

---

# Example Dataset

| House Size | Bedrooms | Price (₹ Lakhs) |
|------------|----------|----------------:|
| 900 | 2 | 45 |
| 950 | 2 | 48 |
| 1200 | 3 | 60 |
| 1500 | 3 | 72 |
| 1800 | 4 | 90 |

Goal:

Predict

```text
House Price
```

---

# How Does a Regression Tree Split?

The process is almost identical to a Classification Tree.

```text
Entire Dataset

↓

Try Every Feature

↓

Try Every Threshold

↓

Evaluate Split

↓

Choose Best Split

↓

Repeat Recursively
```

The only difference is:

Instead of minimizing **Entropy** or **Gini**,

the algorithm minimizes the prediction error.

---

# What is Stored in a Leaf?

Classification Tree

```text
Leaf

Spam
```

Regression Tree

```text
Leaf

₹52,40,000
```

The prediction stored in a leaf is usually the **average (mean)** of all target values reaching that leaf.

---

# Why the Mean?

Suppose a leaf contains

| House Price |
|-------------|
| ₹48L |
| ₹50L |
| ₹52L |
| ₹54L |

Mean

```text
(48+50+52+54)/4

=

₹51 Lakhs
```

The leaf stores

```text
₹51 Lakhs
```

Every new sample reaching this leaf will receive that prediction.

---

# Splitting Criterion for Regression

Instead of Entropy,

Regression Trees measure how spread out the target values are.

Common criteria include:

- Mean Squared Error (MSE)
- Friedman MSE
- Mean Absolute Error (MAE)

The algorithm selects the split that reduces prediction error the most.

---

# Mean Squared Error (MSE)

The most common regression criterion.

Formula

```text
MSE

=

Σ(y−ŷ)²

/

N
```

where

- y = actual value
- ŷ = predicted value

Smaller MSE means better predictions.

---

# Variance Reduction

A good split creates child nodes where target values are more similar.

Example

Before Split

```text
40

45

80

85
```

Very spread out.

After Split

Left

```text
40

45
```

Right

```text
80

85
```

Each child now has lower variance.

This is a much better split.

Goal

Regression

Goal

```text
Reduce Prediction Error
```

Prediction is simply the average value stored in the leaf.

# Feature Importance

One major advantage of Decision Trees is that they naturally estimate **feature importance**.

---

# What is Feature Importance?

Feature importance measures:

> **How useful each feature was in building the tree.**

Features that consistently produce large impurity reductions receive higher importance scores.

---

# Example

Features

- Age
- Salary
- Education
- Experience

Suppose

Age creates the first split.

Salary creates several important splits.

Education is used once.

Experience is never used.

Importance might be

| Feature | Importance |
|----------|-----------:|
| Salary | 0.45 |
| Age | 0.35 |
| Education | 0.20 |
| Experience | 0.00 |

---

# How is Feature Importance Calculated?

At every split,

the algorithm computes

```text
Impurity Before Split

↓

Impurity After Split

↓

Reduction in Impurity
```

Each feature receives credit for the impurity reductions it creates.

Finally,

all reductions are summed and normalized.

---

# Feature Importance Workflow

```text
Feature

↓

Creates Split

↓

Reduces Impurity

↓

Importance Score Increases

↓

Repeat

↓

Normalize Scores

↓

Final Importance
```

---

# Interpretation

High importance

```text
Feature greatly influences predictions.
```

Low importance

```text
Feature contributes little.
```

Zero importance

```text
Feature was never selected for splitting.
```

---

# Limitations of Feature Importance

Feature importance can be biased toward:

- Continuous features
- Features with many unique values

Therefore,

interpret importance carefully,

especially when features differ greatly in scale or cardinality.

# Summary

A Decision Tree is a supervised learning algorithm that builds a hierarchy of decision rules by recursively splitting the data. During training, it measures impurity using **Entropy** or **Gini Impurity**, selects the best split using **Information Gain** or **Gini reduction**, and continues growing until a stopping criterion is met. To improve generalization, techniques such as **pre-pruning** and **post-pruning** are applied. For regression problems, the same framework is used, but leaves predict numerical values by minimizing prediction error. Decision Trees are easy to interpret, naturally provide feature importance, and form the foundation of powerful ensemble methods such as Random Forests and Gradient Boosting.
