# Tree Terminology

Before learning how a Decision Tree chooses the best split, it's important to understand the components that make up the tree.

A Decision Tree is composed of different types of nodes connected by branches.

Example:

```text
                Age < 30 ?
               /          \
            Yes            No
             |              |
      Student?         Salary > 50K?
       /     \           /        \
     Yes     No       Yes         No
      |       |         |           |
    Buy    Don't      Buy        Don't
```

Each part of this tree has a specific name.

---

# 1. Root Node

## Definition

The **Root Node** is the first node of the Decision Tree.

It contains the **entire training dataset** before any splitting has occurred.

Every prediction starts from the root.

---

## Characteristics

- First node in the tree
- Contains all training samples
- Has no parent
- Usually performs the most important split
- There is only one root node

---

## Example

```text
                Age < 30 ?
```

Here,

```text
Age < 30 ?
```

is the Root Node.

---

## Why is it important?

Since every data point passes through the root node, the algorithm tries to choose the feature that reduces uncertainty the most.

Later, we'll learn that this decision is made using:

- Information Gain
- Gini Impurity
- Entropy

---

# 2. Decision Node (Internal Node)

## Definition

A **Decision Node** is any node that performs another split after the root.

It asks another question to further separate the data.

---

## Example

```text
                Age < 30 ?
               /          \
            Yes            No
             |
         Student?
```

Here,

```text
Student?
```

is a Decision Node.

---

## Characteristics

- Has one parent
- Has one or more children
- Contains a condition
- Splits data into smaller groups

---

A tree can have many decision nodes.

Example:

```text
Root
 │
 ├── Decision
 │
 ├── Decision
 │
 ├── Decision
 │
 └── Leaf
```

---

# 3. Leaf Node (Terminal Node)

## Definition

A **Leaf Node** is the final node of the tree.

No further splitting occurs.

Instead, the model makes its prediction here.

---

## Classification Tree

Leaf stores the predicted class.

Example

```text
Spam
```

or

```text
Not Spam
```

---

## Regression Tree

Leaf stores a numerical value.

Example

```text
₹54,30,000
```

or

```text
Temperature = 28.5°C
```

---

Example Tree

```text
             Age < 30?

          /             \
       Yes              No

     Student?       Salary?

    /       \       /      \

  Buy     Don't   Buy    Don't
```

The four bottom nodes are leaf nodes.

---

# Why Leaf Nodes Are Important

Once a sample reaches a leaf,

the algorithm stops asking questions.

The prediction stored inside that leaf becomes the final prediction.

---

# 4. Parent Node

A node that creates another node is called a **Parent Node**.

Example

```text
          Age < 30?

         /        \

      Student?    Salary?
```

Here,

```text
Age < 30?
```

is the parent of

```text
Student?
```

and

```text
Salary?
```

---

# 5. Child Node

Nodes created from another node are called **Child Nodes**.

Example

```text
Age < 30?
     |
 Student?
```

Student? is the child of Age < 30?

---

# Parent-Child Relationship

```text
          Parent

            |

      ----------------

      |              |

   Child 1        Child 2
```

Every node (except the root) has exactly one parent.

A parent can have multiple children.

---

# 6. Branch

A **Branch** is simply the connection between two nodes.

It represents the outcome of a decision.

Example

```text
Age < 30?

     Yes
      |
 Student?
```

The "Yes" connection is a branch.

Similarly,

```text
Age < 30?

     No
      |
 Salary?
```

is another branch.

---

Branches represent the possible answers to a question.

For binary trees,

there are usually two branches:

```text
Yes

No
```

or

```text
True

False
```

---

# 7. Path

A **Path** is the sequence of branches followed from the root node to a leaf node.

Example

```text
Age < 30?

     Yes

Student?

     No

Don't Buy
```

This entire route is one path.

Another path is

```text
Age < 30?

      No

Salary > 50K?

      Yes

Buy
```

Each path represents one decision rule.

---

# Decision Rule

Every path can be converted into an IF-ELSE rule.

Example

```text
Age < 30

AND

Student = No

↓

Don't Buy
```

This becomes

```python
if age < 30 and student == "No":
    prediction = "Don't Buy"
```

Decision Trees are essentially collections of many IF-ELSE rules.

---

# 8. Depth of a Node

The **Depth** of a node is the number of edges from the root to that node.

Example

```text
              Root

             /    \

          Node    Node

          /

       Leaf
```

Depth values

```text
Root = 0

Children = 1

Leaf = 2
```

---

Example

```text
              Age

          /        \

     Student      Salary

        |

      Buy
```

Depth

```text
Age = 0

Student = 1

Buy = 2
```

---

# 9. Tree Depth (Maximum Depth)

The **Depth of a Tree** is the maximum depth among all nodes.

Example

```text
Depth = 3
```

means the longest path from the root to a leaf contains three edges.

---

Why does it matter?

A shallow tree

```text
Depth = 2
```

may underfit.

A very deep tree

```text
Depth = 20
```

may overfit.

Tree depth controls model complexity.

---

# 10. Level

Level is similar to depth.

Some books define

```text
Root = Level 1
```

instead of

```text
Depth = 0
```

Example

```text
Level 1

Root

Level 2

Children

Level 3

Leaves
```

Just be aware that some resources use **depth** while others use **level**.

---

# 11. Subtree

A **Subtree** is any smaller tree inside the main tree.

Example

```text
             Age

          /      \

      Student    Salary

      /     \

    Buy   Don't
```

The Student portion itself forms a subtree.

Subtrees become important in **pruning**, where parts of the tree may be removed to reduce overfitting.

---

# 12. Sibling Nodes

Nodes sharing the same parent are called **siblings**.

Example

```text
          Age

      /        \

 Student      Salary
```

Student and Salary are siblings.

---

# 13. Internal Node

An **Internal Node** is any node that is not a leaf.

This includes:

- Root Node
- Decision Nodes

Example

```text
Age

Student

Salary
```

All are internal nodes because they continue splitting the data.

---

# How Prediction Happens

Suppose a new customer has

```text
Age = 24

Student = Yes
```

Prediction

```text
Start

↓

Root

↓

Age < 30?

↓

Yes

↓

Student?

↓

Yes

↓

Buy
```

The model simply follows one path until it reaches a leaf.

---

# Important Interview Questions

### Why is the root node important?

Because it contains all training data and performs the first split, which usually has the greatest impact on the overall tree.

---

### What is stored inside a leaf node?

- In classification: the predicted class (or class distribution).
- In regression: a numerical prediction, typically the average target value of the samples in that leaf.

---

### Can a Decision Tree have multiple root nodes?

No.

Every Decision Tree has exactly one root node.

---

### What is the difference between a decision node and a leaf node?

| Decision Node | Leaf Node |
|--------------|-----------|
| Performs another split | Makes the final prediction |
| Has child nodes | Has no children |
| Contains a decision condition | Contains the predicted output |

---

# Key Takeaways

- A Decision Tree begins with a single **Root Node** containing all training data.
- **Decision Nodes** ask questions that split the data into smaller subsets.
- **Leaf Nodes** store the final prediction and do not split further.
- **Branches** connect nodes and represent the outcomes of decisions.
- A **Path** from the root to a leaf corresponds to an interpretable IF-ELSE rule.
- **Depth** measures how far a node is from the root, while **tree depth** controls model complexity.
- **Subtrees** are smaller sections of the tree and are important for pruning techniques.
