# Decision Trees - Part 3: Purity and Impurity

---

# Introduction

Now that we understand the structure of a Decision Tree, the next question is:

> **How does the algorithm decide where to split the data?**

A Decision Tree does **not** choose a feature randomly.

Instead, it tries to find the split that creates the **purest possible groups**.

To understand this, we first need to understand two fundamental concepts:

- Purity
- Impurity

These are the foundation of Entropy, Gini Impurity, and Information Gain.

---

# Why Does a Decision Tree Split the Data?

Imagine you're trying to separate apples and oranges.

Initially, they are mixed together.

```text
🍎 🍊 🍎 🍊 🍊 🍎 🍎 🍊
```

This mixture is difficult to classify.

Instead, if we separate them into two baskets:

```text
Basket 1

🍎 🍎 🍎 🍎

Basket 2

🍊 🍊 🍊 🍊
```

Now the classification becomes very easy.

This is exactly what a Decision Tree tries to achieve.

Instead of separating fruits, it separates **data points** belonging to different classes.

---

# Goal of a Decision Tree

The goal of every split is to make the resulting groups as **homogeneous** (similar) as possible.

Another way to say this is:

> A good split produces **pure nodes**.

---

# What is a Pure Node?

## Definition

A **Pure Node** is a node where **all samples belong to the same class**.

There is no uncertainty.

---

## Example 1

Suppose we are classifying emails.

```text
Spam

Spam

Spam

Spam

Spam
```

Every email belongs to the Spam class.

This node is **100% pure**.

---

## Example 2

```text
Not Spam

Not Spam

Not Spam

Not Spam
```

Again,

every sample belongs to the same class.

This node is also pure.

---

# Characteristics of a Pure Node

A pure node has:

- Only one class
- No ambiguity
- No uncertainty
- No need for further splitting
- Usually becomes a leaf node

---

# What is an Impure Node?

## Definition

An **Impure Node** contains samples from **multiple classes**.

The classes are mixed together.

---

## Example

```text
Spam

Spam

Not Spam

Spam

Not Spam
```

Now,

both classes exist.

The algorithm is uncertain.

This node is impure.

---

# Characteristics of an Impure Node

An impure node has:

- Multiple classes
- High uncertainty
- Difficult prediction
- Needs further splitting

---

# Visual Comparison

## Pure Node

Prediction is obvious.

↓

---

## Impure Node

```text
Spam

Spam

Not Spam

Spam

Not Spam
```

Prediction is uncertain.

The algorithm wants to split this node.

# Another Example

Suppose we're predicting whether a customer buys a laptop.

Node A

```text
Buy

Buy

Buy

Buy

Buy
```

This node is pure.

---

Node B

```text
Buy

Buy

Don't Buy

Buy

Don't Buy
```

This node is impure.

# How Does the Tree Think?

Imagine the algorithm asking itself:

```text
Current node

↓

Are all samples the same?

↓

Yes

↓

Stop splitting.

↓

Make this a leaf.
```

Otherwise,

```text
Current node

↓

Classes are mixed.

↓

Need another split.

↓

Search for the best feature.
```

# Why Not Keep Splitting Forever?

Imagine we continue splitting until every leaf contains exactly one training sample.

```text
Leaf 1

Yes
```

```text
Leaf 2

No
```

```text
Leaf 3

Yes
```

Every leaf is perfectly pure.

Training accuracy becomes **100%**.

But this is a problem.

The model has memorized the training data instead of learning general patterns.

This is called **overfitting**.

Later, we'll learn techniques such as:

- Maximum Depth
- Minimum Samples Leaf
- Cost Complexity Pruning

to prevent this.

# Measuring Purity

Humans can visually identify whether a node is pure or impure.

A computer needs a numerical measure.

It asks:

> **How impure is this node?**

To answer this, Decision Trees use mathematical measures called **impurity metrics**.

The most common are:

- Entropy
- Gini Impurity

These metrics assign a number to each node.

A lower value means the node is more pure.

---

# Intuition Behind Impurity Scores

Suppose we have three nodes.

### Node A

```text
Yes

Yes

Yes

Yes
```

Impurity = Very Low

---

### Node B

```text
Yes

Yes

Yes

No
```

Impurity = Medium

---

### Node C

```text
Yes

No

Yes

No
```

Impurity = High

---

As the classes become more mixed,

the impurity increases.

---

# Everyday Analogy

Imagine a classroom.

### Classroom A

Every student scored 95%.

All students perform similarly.

Very little uncertainty.

Equivalent to a **pure node**.

---

### Classroom B

Some students scored 95%.

Some scored 70%.

Some scored 30%.

Performance varies greatly.

Much more uncertainty.

Equivalent to an **impure node**.

---

# Relationship Between Purity and Splitting

A Decision Tree repeatedly performs this process:

```text
Current Node

↓

Measure Impurity

↓

Try Different Splits

↓

Choose the Split

↓

Impurity Decreases

↓

Repeat
```

The algorithm stops when:

- Nodes become sufficiently pure.
- A stopping criterion (e.g., max depth or minimum samples) is reached.
- No split can further reduce impurity.

---

# Important Interview Questions

### Why does a Decision Tree split data?

To reduce impurity and create groups that are as homogeneous (pure) as possible.

---

### What is a pure node?

A node in which every sample belongs to the same class.

---

### What is an impure node?

A node containing samples from two or more classes.

---

### Why are pure nodes desirable?

Because they lead to confident predictions and usually require no further splitting.

---

### Does a Decision Tree measure purity directly?

No.

It measures **impurity** using metrics such as:

- Entropy
- Gini Impurity

Lower impurity means higher purity.

---

# Key Takeaways

- The primary objective of a Decision Tree is to reduce impurity at every split.
- A **pure node** contains samples from only one class and usually becomes a leaf.
- An **impure node** contains mixed classes and requires further splitting.
- Good splits create child nodes that are more homogeneous than the parent node.
- Impurity is quantified using mathematical metrics, most commonly **Entropy** and **Gini Impurity**.
- Understanding purity and impurity is essential before studying Entropy, Gini, and Information Gain.
