# Decision Trees - Part 4.1: Entropy (Foundation of Decision Tree Splitting)

---

# Introduction

In the previous chapter, we learned that a Decision Tree aims to create **pure nodes** by repeatedly splitting the dataset.

However, a computer cannot simply "look" at a node and determine whether it is pure or impure.

Humans can easily observe:

```text
Spam
Spam
Spam
Spam
```

and immediately recognize that the node is pure.

Similarly,

```text
Spam
Not Spam
Spam
Not Spam
```

is clearly impure.

A computer, however, requires a **numerical measure** of impurity.

This leads to the concept of **impurity metrics**.

The two most common impurity metrics are:

- Entropy
- Gini Impurity

In this chapter, we begin with **Entropy**, one of the most fundamental concepts in Decision Trees.

---

# Why Do We Need a Mathematical Measure?

Suppose a Decision Tree is considering two possible splits.

## Split A

```text
Left Node

Spam
Spam
Spam

Right Node

Not Spam
Not Spam
Not Spam
```

Both nodes are perfectly pure.

---

## Split B

```text
Left Node

Spam
Spam
Not Spam

Right Node

Spam
Not Spam
Not Spam
```

These nodes are still mixed.

---

A human immediately knows that **Split A is better**.

A computer cannot make this judgment visually.

Instead, it assigns a numerical score to each split and chooses the one with the **lowest impurity** (or equivalently, the highest reduction in impurity).

This score is computed using metrics such as **Entropy** or **Gini Impurity**.

---

# Before Learning Entropy: Probability Refresher

Entropy is calculated using probabilities.

Therefore, we first need to understand how to compute the probability of each class within a node.

Suppose a node contains:

```text
Spam
Spam
Spam
Not Spam
Not Spam
```

Total samples = **5**

Spam = **3**

Not Spam = **2**

The probability of each class is:

```text
P(Spam) = 3 / 5 = 0.6

P(Not Spam) = 2 / 5 = 0.4
```

Notice that:

```text
0.6 + 0.4 = 1
```

The probabilities of all classes in a node always sum to **1**.

---

# Another Example

Suppose a node contains:

```text
Yes
Yes
Yes
Yes
No
```

Total samples = **5**

Probability:

```text
P(Yes) = 4/5 = 0.8

P(No) = 1/5 = 0.2
```

Again,

```text
0.8 + 0.2 = 1
```

These probabilities will be used in the Entropy formula.

---

# What is Entropy?

## Definition

**Entropy** is a measure of the **uncertainty**, **disorder**, or **impurity** within a node.

In Decision Trees:

- High Entropy → High impurity (mixed classes)
- Low Entropy → Low impurity (mostly one class)
- Zero Entropy → Completely pure node

The Decision Tree tries to **reduce Entropy** after every split.

---

# Everyday Intuition

Imagine a bag of balls.

### Bag A

```text
⚪ ⚪ ⚪ ⚪ ⚪
```

Every ball is white.

If you randomly pick one ball, you are certain it will be white.

There is **no uncertainty**.

Entropy is **0**.

---

### Bag B

```text
⚪ ⚫ ⚪ ⚫ ⚪ ⚫
```

Half white, half black.

Now you cannot predict the color with certainty.

This bag has **high uncertainty**.

Entropy is **maximum**.

---

### Bag C

```text
⚪ ⚪ ⚪ ⚪ ⚫
```

Mostly white.

There is some uncertainty, but less than Bag B.

Entropy is **moderate**.

---

# The Entropy Formula

For a classification problem with \(n\) classes:

```text
Entropy(S) = - Σ [ P(i) × log₂(P(i)) ]
```

where:

- **S** = current node
- **P(i)** = probability of class *i*
- **log₂** = logarithm with base 2
- **Σ** = sum over all classes

---

# Binary Classification Formula

For two classes:

```text
Entropy(S)

=

− P(Positive) log₂(P(Positive))

− P(Negative) log₂(P(Negative))
```

This is the formula you'll use most often.

---

# Why is There a Negative Sign?

The logarithm of a probability (between 0 and 1) is negative.

For example:

```text
log₂(0.5) = -1

log₂(0.25) = -2
```

Without the negative sign, entropy would be negative.

The leading minus sign ensures that entropy is always non-negative.

---

# Special Logarithm Convention

When a class has probability **0**, the term:

```text
0 × log₂(0)
```

is mathematically undefined.

In information theory, we define:

```text
0 × log₂(0) = 0
```

This convention allows entropy to remain well-defined even when a class is absent.

---

# Example 1: Perfectly Pure Node

Suppose a node contains:

```text
Spam
Spam
Spam
Spam
```

Probability:

```text
P(Spam) = 1

P(Not Spam) = 0
```

Entropy:

```text
Entropy

=

-(1 × log₂1)

-(0 × log₂0)

=

0
```

This node is perfectly pure.

---

# Example 2: Maximum Uncertainty

Suppose:

```text
Spam

Spam

Not Spam

Not Spam
```

Probability:

```text
P(Spam)=0.5

P(Not Spam)=0.5
```

Entropy:

```text
= -(0.5 log₂0.5)

-(0.5 log₂0.5)

= -(0.5 × -1)

-(0.5 × -1)

= 0.5 + 0.5

= 1
```

Entropy is **1**, which is the maximum for binary classification.

---

# Example 3: Mostly One Class

Node:

```text
Spam

Spam

Spam

Not Spam
```

Probability:

```text
P(Spam)=0.75

P(Not Spam)=0.25
```

Using:

```text
log₂0.75 ≈ -0.415

log₂0.25 = -2
```

Entropy:

```text
= -(0.75 × -0.415)

-(0.25 × -2)

= 0.311

+0.500

≈0.811
```

Entropy is approximately **0.81**.

This node is less impure than the 50–50 split.

# Interpreting Entropy

| Entropy | Interpretation |
|----------|----------------|
| 0 | Completely pure |
| 0.2 | Very low uncertainty |
| 0.5 | Moderately pure |
| 0.8 | Quite mixed |
| 1 | Maximum uncertainty (binary) |

As entropy increases, uncertainty increases.

As entropy decreases, the node becomes more homogeneous.

---

# Entropy Curve

```text
Entropy

1.0 |                 ●
    |              ●     ●
0.8 |           ●           ●
    |        ●               ●
0.6 |      ●                   ●
    |    ●                       ●
0.4 |  ●                           ●
    | ●                             ●
0.2 |●                               ●
    |
0.0 +---------------------------------------> P(Class 1)

    0      0.25      0.5      0.75       1
```

The curve is symmetric.

Entropy is highest when both classes are equally likely.

---

# Why is the Maximum at 50–50?

Imagine tossing a fair coin.

```text
Heads = 50%

Tails = 50%
```

Before tossing, you have maximum uncertainty.

Now imagine a biased coin:

```text
Heads = 99%

Tails = 1%
```

You can almost always guess "Heads."

Uncertainty is much lower.

This is exactly what entropy measures.

---

# Entropy in Multiclass Classification

Entropy also works for more than two classes.

Example:

```text
Cat

Dog

Bird
```

Probabilities:

```text
Cat = 1/3

Dog = 1/3

Bird = 1/3
```

The entropy formula simply sums over all classes.

As the number of equally likely classes increases, the maximum possible entropy also increases.

---

# Important Properties of Entropy

- Entropy is always greater than or equal to 0.
- Entropy equals 0 for a perfectly pure node.
- Higher entropy means greater uncertainty.
- Entropy depends only on class probabilities, not feature values.
- The Decision Tree seeks splits that reduce entropy.

---

# Common Misconceptions

### Misconception 1

**Entropy measures accuracy.**

❌ Incorrect.

Entropy measures **impurity**, not predictive accuracy.

---

### Misconception 2

**A node with more samples always has higher entropy.**

❌ Incorrect.

Entropy depends on the **distribution of classes**, not the number of samples.

For example:

```text
100 Spam
```

Entropy = 0

```text
2 Spam

2 Not Spam
```

Entropy = 1

---

### Misconception 3

**Entropy is always between 0 and 1.**

❌ Only for binary classification.

For multiclass problems, the maximum entropy is greater than 1.

---

# Key Takeaways

- Entropy measures the uncertainty or impurity of a node.
- Pure nodes have an entropy of **0**.
- Binary entropy reaches its maximum value of **1** when the classes are evenly split (50–50).
- Entropy depends solely on the class probabilities within a node.
- A Decision Tree aims to reduce entropy with every split.
- Entropy alone does not choose the best split; it only evaluates the impurity of a node.

# Decision Trees - Part 4.2: Information Gain (How a Decision Tree Chooses the Best Split)

---

# Introduction

In the previous chapter, we learned that **Entropy** measures the impurity or uncertainty of a node.

However, knowing the entropy of a node is **not enough** to build a Decision Tree.

The algorithm must answer a more important question:

> **Among all possible splits, which one should be chosen?**

This is where **Information Gain** comes into the picture.

Information Gain tells us **how much impurity (entropy) decreases after a split**.

The Decision Tree always chooses the split that produces the **highest Information Gain**.

---

# The Main Goal of a Split

Suppose the root node contains:

```text
Spam

Spam

Spam

Not Spam

Not Spam

Not Spam
```

This node is mixed.

The Decision Tree now tries different features and thresholds.

For each possible split, it asks:

> **"Did this split reduce the uncertainty?"**

If yes,

How much?

The answer is **Information Gain**.

---

# Intuition

Imagine you are solving a mystery.

Initially,

you know very little.

```text
Uncertainty = High
```

Then someone gives you useful information.

Now,

your uncertainty decreases.

The amount by which uncertainty decreases is called **Information Gain**.

Decision Trees work in exactly the same way.

Every split should reduce uncertainty.

---

# Information Gain Definition

Information Gain is the **reduction in entropy after splitting a node**.

Mathematically,

```text
Information Gain

=

Entropy of Parent

−

Weighted Entropy of Children
```

If the child nodes are much purer than the parent,

Information Gain will be high.

---

# Information Gain Formula

```text
IG

=

Entropy(Parent)

−

Weighted Entropy(Children)
```

where

```text
Weighted Entropy

=

(n₁ / N) × Entropy(Child₁)

+

(n₂ / N) × Entropy(Child₂)

+

...
```

Here,

- **N** = total samples in the parent node
- **n₁, n₂** = samples in each child node

The weights ensure that larger child nodes contribute more to the final impurity.

---

# Why Weighted Entropy?

Suppose a split creates:

```text
Left Child = 95 samples

Right Child = 5 samples
```

Clearly,

the larger child should influence the impurity more than the smaller child.

Therefore,

we compute a **weighted average** instead of a simple average.

---

# Example of Weighted Average

Suppose

Child A

```text
90 students
```

Average marks

```text
80
```

Child B

```text
10 students
```

Average marks

```text
40
```

Overall average is **not**

```text
(80+40)/2 = 60
```

because the class sizes are different.

Instead,

```text
(90×80 + 10×40)

/

100

=

76
```

Decision Trees use the same idea when combining child entropies.

For a correctly implemented Decision Tree,

**No.**

A valid split cannot increase entropy beyond the weighted entropy of the parent.

Therefore,

```text
Information Gain ≥ 0
```

# Why Is It Called a Greedy Algorithm?

Decision Trees choose the **best split at the current node**.

They do **not** check whether a slightly worse split now could lead to a better overall tree later.

This approach is called a **Greedy Algorithm**.

It makes the best local decision at each step.

# Advantages of Information Gain

- Easy to interpret
- Produces highly informative splits
- Works well for classification problems
- Based on solid information theory
- Naturally favors purer child nodes

---

# Limitations of Information Gain

Information Gain has one important drawback.

It tends to favor features with **many unique values**.

Example

```text
Student ID
```

Every student has a unique ID.

Splitting on Student ID creates perfectly pure nodes.

Training accuracy becomes 100%.

But this is useless for prediction.

This bias is one reason why algorithms such as **C4.5** use **Gain Ratio** instead of raw Information Gain.

---

### What does a high Information Gain indicate?

### Why is weighted entropy used?

Because child nodes may contain different numbers of samples, and larger nodes should contribute proportionally more to the impurity calculation.

---

### Why are Decision Trees called greedy algorithms?

Because they choose the best split at the current node without considering whether that choice leads to the globally optimal tree.

### Can Information Gain be negative?

No.

# Key Takeaways

- Information Gain measures how much entropy decreases after a split.
- It is calculated as the parent's entropy minus the weighted entropy of the child nodes.
- The Decision Tree evaluates all candidate splits and selects the one with the highest Information Gain.
- Weighted entropy accounts for differences in the sizes of the child nodes.
- Information Gain is a greedy criterion—it optimizes each split independently.
- Although powerful, Information Gain can be biased toward features with many unique values.

# Decision Trees - Part 4.3: Gini Impurity, Entropy vs Gini, and the Complete Splitting Algorithm

---

# Introduction

So far, we have learned:

- **Entropy** measures the impurity of a node.
- **Information Gain** measures how much impurity decreases after a split.

However, Entropy is **not the only impurity measure** used in Decision Trees.

Many Decision Tree algorithms—including the **CART (Classification and Regression Trees)** algorithm used by **scikit-learn's `DecisionTreeClassifier`**—use another metric called **Gini Impurity**.

Both Entropy and Gini serve the same purpose:

> **Measure how mixed the classes are inside a node.**

The only difference lies in **how they calculate impurity**.

---

# What is Gini Impurity?

## Definition

**Gini Impurity** measures the probability that a randomly selected sample would be **incorrectly classified** if it were randomly assigned a class according to the class distribution in that node.

Although this definition sounds mathematical, the intuition is simple.

- If a node contains only one class, there is **no chance of making a mistake**.
- If a node contains many mixed classes, the chance of making a mistake is higher.

Therefore,

- Low Gini → Pure node
- High Gini → Impure node

---

# Everyday Intuition

Imagine a bag containing balls.

## Bag A

```text
⚪ ⚪ ⚪ ⚪ ⚪
```

Every ball is white.

If you randomly guess "White,"

you will always be correct.

Mistake probability = 0

Gini = 0

---

## Bag B

```text
⚪ ⚫ ⚪ ⚫ ⚪ ⚫
```

Half white

Half black

Now,

random guessing becomes much harder.

Mistake probability increases.

Gini increases.

---

# Gini Formula

For **n** classes,

```text
Gini

=

1

-

Σ(P(i)²)
```

where

- P(i) = probability of class i

---

# Binary Classification Formula

For two classes,

```text
Gini

=

1

-

[P(Yes)² + P(No)²]
```

This is the formula you'll use most often.

---

# Example 1: Pure Node

Suppose

```text
Spam

Spam

Spam

Spam
```

Probability

```text
P(Spam)=1

P(Not Spam)=0
```

Gini

```text
=

1

-

(1²+0²)

=

0
```

Perfectly pure.

---

# Example 2: Equal Distribution

Suppose

```text
Spam

Spam

Not Spam

Not Spam
```

Probability

```text
0.5

0.5
```

Gini

```text
=

1

-

(0.5²+0.5²)

=

1

-

(0.25+0.25)

=

0.5
```

Maximum impurity for binary classification.

---

# Example 3: Mostly One Class

Suppose

```text
Spam

Spam

Spam

Not Spam
```

Probability

```text
P(Spam)=0.75

P(Not Spam)=0.25
```

Gini

```text
=

1

-

(0.75²+0.25²)

=

1

-

(0.5625+0.0625)

=

0.375
```

Lower than 0.5,

which indicates a purer node.

---

# Range of Gini

For binary classification:

| Gini | Meaning |
|------:|---------|
| 0 | Completely pure |
| 0.1 | Very pure |
| 0.25 | Fairly pure |
| 0.40 | Mixed |
| 0.50 | Maximum impurity |

Unlike Entropy,

the maximum Gini value is

```text
0.5
```

for binary classification.

---

# Gini Curve

```text
Gini

0.5 |                ●
    |             ●     ●
0.4 |          ●           ●
    |       ●               ●
0.3 |     ●                   ●
    |   ●                       ●
0.2 | ●                           ●
    |●                             ●
0.1 |
    |
0.0 +-------------------------------------> P(Class 1)

    0      0.25      0.5      0.75      1
```

The curve is also symmetric.

Maximum impurity occurs when both classes are equally likely.

---

# Why Does scikit-learn Use Gini by Default?

The default parameter in `DecisionTreeClassifier` is

```python
criterion="gini"
```

Reasons:

- Simpler computation
- No logarithms
- Faster training
- Similar accuracy to Entropy
- Efficient for large datasets

In practice,

Entropy and Gini often produce very similar trees.

---

# How CART Uses Gini

The **CART (Classification and Regression Trees)** algorithm:

1. Computes the Gini impurity of the parent node.
2. Tries every possible split.
3. Computes the weighted Gini impurity of the child nodes.
4. Chooses the split with the **lowest weighted Gini**.
5. Repeats the process recursively.

Unlike ID3, CART always creates **binary splits**.

---

# Entropy vs Gini

Both metrics measure impurity.

However,

they differ in several ways.

| Feature | Entropy | Gini Impurity |
|---------|---------|---------------|
| Formula | −Σ P log₂P | 1 − ΣP² |
| Uses logarithm | Yes | No |
| Computational cost | Higher | Lower |
| Speed | Slightly slower | Slightly faster |
| Maximum value (Binary) | 1 | 0.5 |
| Pure node | 0 | 0 |
| Default in scikit-learn | No | Yes |
| Used in ID3 | Yes | No |
| Used in CART | No | Yes |

---

# Which One is Better?

There is no universal winner.

### Entropy

Advantages

- Strong theoretical foundation in Information Theory.
- Sometimes produces slightly more balanced trees.
- Useful for understanding Information Gain.

---

### Gini

Advantages

- Faster computation.
- Simpler formula.
- Preferred by CART.
- Default choice in scikit-learn.

---

In practice,

the difference in predictive performance is usually very small.

# Common Mistakes

## Mistake 1

Thinking Entropy and Gini produce completely different trees.

Usually,

they produce very similar trees.

---

## Mistake 2

Assuming lower Information Gain is better.

Incorrect.

Higher Information Gain means a better split.

---

## Mistake 3

Comparing Entropy values directly with Gini values.

These metrics have different scales.

Only compare values within the same metric.

---

## Mistake 4

Believing Gini works only for binary classification.

Incorrect.

Gini also supports multiclass classification.

# Important Interview Questions

### Why does scikit-learn use Gini by default?

Because it is computationally faster, does not require logarithmic calculations, and generally performs similarly to Entropy.

---

### What is the difference between Entropy and Gini?

Both measure impurity, but Entropy uses logarithms while Gini uses squared probabilities.

---

### Which impurity measure is faster?

Gini Impurity.

---

### Can Gini be used for multiclass classification?

Yes.

---

### What is the maximum Gini value for binary classification?

0.5

---

### Does the Decision Tree check every possible feature?

Yes.

For numerical features,

it also evaluates many possible thresholds before selecting the best split.

# Cheat Sheet

| Concept | Purpose |
|----------|---------|
| Entropy | Measures uncertainty using information theory |
| Information Gain | Measures reduction in entropy after a split |
| Gini Impurity | Measures impurity using squared probabilities |
| Best Split | Highest Information Gain or Lowest Weighted Gini |
| Pure Node | Entropy = 0, Gini = 0 |
| Binary Entropy Range | 0 → 1 |
| Binary Gini Range | 0 → 0.5 |
| CART | Uses Gini |
| ID3 | Uses Entropy |

# Key Takeaways

- Gini Impurity is an alternative to Entropy for measuring node impurity.
- CART, the algorithm behind `DecisionTreeClassifier` in scikit-learn, uses Gini by default.
- Entropy and Gini usually produce similar Decision Trees, although Gini is computationally simpler.
- A Decision Tree evaluates every candidate split, computes impurity, and greedily selects the best one.
- Tree construction continues recursively until a stopping criterion is met.
- Understanding Entropy, Information Gain, and Gini provides the mathematical foundation for how Decision Trees learn from data.
