👉 Each point = different threshold

### ✅ Code to Plot ROC

## 🔥 Step 2: Find Optimal Threshold (ROC)

### 🎯 Method: Maximize TPR - FPR (Youden’s Index)
``` python
import numpy as np

optimal_idx = np.argmax(tpr - fpr)
optimal_threshold = thresholds[optimal_idx]

print("Optimal Threshold (ROC):", optimal_threshold)
```
### 🔍 What this does

## Maximizes: TPR - FPR

### 👉 Meaning:

## High recall
## Low false positives

# 🔹 Visual Intuition

## 🔥 Step 3: Precision-Recall Curve (BETTER for imbalanced data)

1. 👉 Your spam dataset is imbalanced
2. 👉 PR curve is more useful than ROC

✅ Code
``` python
from sklearn.metrics import precision_recall_curve

precision, recall, thresholds = precision_recall_curve(y_test, lr_probs)

plt.plot(recall, precision)
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve")
plt.show()
🔥 Step 4: Choose Optimal Threshold (PR Curve)
🎯 Method: Maximize F1 Score
f1_scores = 2 * (precision * recall) / (precision + recall + 1e-10)

optimal_idx = np.argmax(f1_scores)
optimal_threshold = thresholds[optimal_idx]

print("Optimal Threshold (PR):", optimal_threshold)
```

### 🔍 Why this works

F1 balances: 
1. Precision (avoid false positives)
2. Recall (catch positives)


## 🔥 When to use what
1. Situation	Use Balanced data -	ROC
2. Imbalanced data (like spam)	- PR curve ✅
3. Want best tradeoff	- F1 score
4. Want fewer false positives	 - High threshold
5. Want fewer false negatives	- Low threshold
🔥 Apply the threshold
```python
custom_preds = (lr_probs >= optimal_threshold).astype(int)

from sklearn.metrics import classification_report
print(classification_report(y_test, custom_preds)) 
```
