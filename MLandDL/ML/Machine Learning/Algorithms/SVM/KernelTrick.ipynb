# Interview Questions

### Q1. Why do we need kernels?

Because many real-world datasets are not linearly separable. Kernels allow SVM to find linear separators in a transformed higher-dimensional feature space without explicitly computing that transformation.

---

### Q2. What is the Kernel Trick?

The Kernel Trick computes the dot product in a transformed feature space directly using a kernel function, avoiding the need to explicitly generate the transformed features.

---

### Q3. Why is LinearSVC preferred for text classification?

Text data represented by TF-IDF is already high-dimensional and often linearly separable. A linear kernel is computationally efficient and typically performs as well as or better than non-linear kernels for this type of data.

---

### Q4. What does the gamma parameter control in the RBF kernel?

Gamma determines the radius of influence of each training sample. A small gamma creates smoother decision boundaries, while a large gamma creates highly localized and potentially overfit decision boundaries.

---

### Q5. Which kernel should you try first?

- **Linear** for text data or when the number of features is very large.
- **RBF** as a strong default for non-linear tabular datasets.
- **Polynomial** only when you have reason to believe polynomial feature interactions are important.

---

# Cheat Sheet

- **Linear Kernel:** \(K(x,y)=x^Ty\)
- **Polynomial Kernel:** \(K(x,y)=(x^Ty+c)^d\)
- **RBF Kernel:** \(K(x,y)=\exp(-\gamma\|x-y\|^2)\)
- **Kernel Trick:** Compute dot products in high-dimensional space without explicitly mapping the features.
- **Small Gamma:** Smooth boundary, lower variance, possible underfitting.
- **Large Gamma:** Complex boundary, higher variance, possible overfitting.
- **LinearSVC:** Best choice for high-dimensional sparse text data (e.g., TF-IDF).
