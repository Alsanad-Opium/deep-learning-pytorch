L2 Regularization (Ridge) Cheat Sheet
1. Definition

L2 Regularization adds a penalty to the loss function equal to the sum of squared weights.

It discourages large weights and helps prevent overfitting.

Also called:

Ridge Regression (in linear models)
Weight decay (in neural networks)

3. Key Intuition
Penalizes large weights heavily
Encourages small but non-zero weights
Smoothens model complexity
4. Effect on Model
Aspect	Effect
Weights	Shrink towards 0
Overfitting	Reduced
Bias	Slightly increases
Variance	Decreases
Model complexity	Reduced
5. Gradient Update Rule

For gradient descent:

w:=w−η(∇J+2λw)

Where:

η = learning rate
2λw = regularization term
6. Geometric Intuition
L2 constraint forms a circle (or sphere in higher dimensions)
Solution prefers points close to origin
Smooth shrinkage instead of eliminating weights
7. L2 vs L1 Regularization
Feature	L2 (Ridge)	L1 (Lasso)
Penalty	Sum of squares	Sum of absolute values
Effect	Shrinks weights	Makes some weights 0
Feature selection	❌ No	✅ Yes
Geometry	Smooth circle	Sharp corners (diamond)
8. When to Use L2

Use L2 when:

You want all features but smaller influence
Data is noisy
Features are correlated
You want stable models
9. Python (Scikit-learn)
from sklearn.linear_model import Ridge

model = Ridge(alpha=1.0)  # alpha = λ
model.fit(X_train, y_train)
10. Hyperparameter (λ / alpha)
Small λ → almost no regularization → overfitting risk
Large λ → strong regularization → underfitting risk
11. Key Takeaway

L2 regularization = “Keep all features, but make the model less confident about any single one.”
