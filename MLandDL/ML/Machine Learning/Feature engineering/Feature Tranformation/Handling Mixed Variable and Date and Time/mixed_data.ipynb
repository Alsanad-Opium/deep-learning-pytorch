Eg 12-march-2027, B34, C54

A data like this can be difficult to deal with in this notebook we will be dealin gwiht this kind of data

```python
import pandas as pd 
import numpy as np 
```

```python
df  = pd.read_csv('titanic.csv')
df.head()
```

```python
df['number'].unique()
```

## In the above dataset we have two mix data columns that are ticket and number 

```python
## converting the number column to as follows  numeric value-> number_numeric  Categotical value -> number_categorical


df['number_numeric'] = pd.to_numeric(df['number'], errors = 'coerce', downcast = 'integer')

# what this peice of code will do is it convert the number col numeric value and add it to the the new column number numeric and for it is not bale to convert it would through an error but that would be ignored and replace witht NaN 

df.head()
```

```python
## Now for converting the categorical value in the number col to categorical value and add it to the  number_categorical col

df['number_categorical'] = np.where(df['number_numeric'].isnull(),df.number,np.nan)


# syntax of the np.where is as follows   
# np.where(condition, value_if_true, value_if_false)

# What the above code will do si check for the valeu in the number_numeric col if it is null and if it returns true it will add the original valeu from the number col and add it to the number categorical col and if the returns false it will return NaN

## Eg  
# Original (number)	number_numerical	number_categorical
# "123"	                123	                    NaN
# "45"	                 45	                    NaN
# "abc"	                NaN	                    "abc"
# "10xyz"	            NaN	                    "10xyz"

df.head()
```

```python
# now we wil deal with the cabin col 


df.head()
```

```python
df['Cabin'].unique()
```

```python
## this hence we cannnot apply encoding on it as it would result in too many columnand it would be hard ofr us to interpret

df['cabin_number'] = df['Cabin'].str.extract(r'(\d+)')
# this will extract the number from the cabin col we have used regex pattern \d+ which will only extract numbers from the cabin col and add it to new col

df.head()
```

```python
df['cabin_cate'] = df['Cabin'].str[0]
# this will extract the first letter of the cabin col  As for our use case the Cabin col values has the letters are the first letter of the col followed by the number 

df.head()
```

```python
## Now comes the final of handling the mixed variable which is the ticket col 

df['Ticket'].value_counts()
```

```python
df['ticket_number'] = df['Ticket'].apply(lambda x: x.split()[-1])
df['ticket_number'] = pd.to_numeric(df.ticket_number,errors = 'coerce',downcast = 'integer')

df.ticket_number.value_counts()
```

```python
df['ticket_number_method2'] = df['Ticket'].str.extract(r'(\d+)')
df.ticket_number_method2.value_counts()
```

```python
# we will be using the First methods as it is more accurate as the second method will only extract the number from the ticket col and it would not consider the case where there is no number in the ticket col and it would return NaN for those cases which is not what we want as we want to keep the original value for those cases where there is no number in the ticket col
df.drop(['ticket_number_method2'],axis =1 , inplace = True)
```

```python
df["ticket_categorical"] = df['Ticket'].apply(lambda x: x.split()[0])
df.ticket_categorical = np.where(df.ticket_categorical.str.isnumeric(),np.nan,df.ticket_categorical)

df.head()
```
