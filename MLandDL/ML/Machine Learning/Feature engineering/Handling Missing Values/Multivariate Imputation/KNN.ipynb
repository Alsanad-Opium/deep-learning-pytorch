```python
import pandas as pd 
import numpy as np 
from sklearn.impute import SimpleImputer, KNNImputer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('Titanic-Dataset.csv')
df.head(4)
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
df = df[['Age', 'Fare','Survived','Pclass']]
df.head(4)
```

```python
x_train,x_test,y_train,y_test = train_test_split(df.drop('Survived',axis=1), df['Survived'],test_size=0.2, random_state=42)
```

```python
knn = KNNImputer(n_neighbors= 2)
x_train_knn = knn.fit_transform(x_train)
x_test_knn = knn.transform(x_test)
```

```python
x_train_knn.shape
x_train_knn[:5]
```

```python
lr = LogisticRegression()

lr.fit(x_train_knn, y_train)
y_pred = lr.predict(x_test_knn)
accuracy_score(y_test,y_pred)
```

```python
si = SimpleImputer(strategy='median')
x_train_si = si.fit_transform(x_train)
x_test_si = si.transform(x_test)
x_train_si[:5]
```

```python
lr.fit(x_train_si,y_train)
y_pred_si = lr.predict(x_test_si)
accuracy_score(y_test,y_pred_si)
```

```python
print(f'Knn accuracy :{accuracy_score(y_test,y_pred)}')
print(f'Simple IMputer accuracy :{accuracy_score(y_test,y_pred_si)}')
```
